```python
"""BLOCK 2 - Load Synthesis and Analysis"""

# =========================
# BLOCK 2A: LOAD CURRENT SUBMISSION + AUTO-MAP KOBO COLUMNS
# =========================

import os
import re
import json
import math
from datetime import datetime
from typing import Dict, List, Tuple, Any

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

PROJECTS_ROOT = "/content/drive/My Drive/2026/tenet/SSOT/RFQI v2/submissions/"
REGISTRY_PATH = os.path.join(PROJECTS_ROOT, "submission_registry.json")

def n(v: Any, default: float = 0.0) -> float:
    if v is None:
        return default
    try:
        if isinstance(v, float) and math.isnan(v):
            return default
    except Exception:
        pass
    try:
        if pd.isna(v):
            return default
    except Exception:
        pass
    try:
        return float(v)
    except Exception:
        return default

def s(v: Any) -> str:
    if v is None:
        return ""
    try:
        if pd.isna(v):
            return ""
    except Exception:
        pass
    return str(v).strip()

def make_json_safe(obj):
    if isinstance(obj, dict):
        return {str(k): make_json_safe(v) for k, v in obj.items()}
    if isinstance(obj, (list, tuple, set)):
        return [make_json_safe(v) for v in obj]
    if isinstance(obj, (pd.Timestamp, datetime)):
        return obj.isoformat()
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.floating):
        return float(obj)
    if isinstance(obj, np.bool_):
        return bool(obj)
    if isinstance(obj, np.ndarray):
        return [make_json_safe(v) for v in obj.tolist()]
    try:
        if pd.isna(obj):
            return None
    except Exception:
        pass
    if isinstance(obj, (str, int, float, bool)) or obj is None:
        return obj
    return str(obj)

def save_json(path, data):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(make_json_safe(data), f, indent=4, ensure_ascii=False)

def load_json(path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)

def norm_text(x: str) -> str:
    x = s(x).lower()
    x = x.replace("&", "and")
    x = re.sub(r"[^a-z0-9]+", " ", x)
    x = re.sub(r"\s+", " ", x).strip()
    return x

registry = load_json(REGISTRY_PATH)
submissions = registry.get("submissions", {})
latest_key = max(submissions.keys(), key=lambda k: submissions[k].get("last_seen_at", ""))
current_entry = submissions[latest_key]

project_root = current_entry["project_root"]
intake_dir = os.path.join(project_root, "00_intake")
load_dir = os.path.join(project_root, "01_load")
os.makedirs(load_dir, exist_ok=True)

raw_state_path = os.path.join(intake_dir, "rfq_raw_state.json")
raw_state = load_json(raw_state_path)
raw_inputs = raw_state.get("raw_inputs", {})
meta = raw_state.get("meta", {})

row_dict = {}
row_dict.update(raw_inputs)
row_dict.update(meta)
df = pd.DataFrame([row_dict])
row = df.iloc[0]

rebuilt_kobo_row_path = os.path.join(load_dir, "rebuilt_kobo_row_for_v1.xlsx")
df.to_excel(rebuilt_kobo_row_path, index=False)

available_columns = list(df.columns)
normalized_column_index = {norm_text(c): c for c in available_columns}

COLUMN_ALIASES = {
    "site_name": [
        "site project name", "site / project name", "site name", "project name"
    ],
    "customer_name": [
        "customer lead name", "customer / lead name", "customer name", "lead name"
    ],
    "supply_phase": [
        "supply phase"
    ],
    "system_priority": [
        "system priority"
    ],
    "bedrooms": [
        "how many bedrooms", "bedrooms"
    ],
    "halls": [
        "how many halls", "how many halls living rooms", "how many halls / living rooms",
        "how many living rooms", "halls", "living rooms"
    ],
    "kitchens": [
        "how many kitchens", "kitchens"
    ],
    "washrooms_override": [
        "washrooms total override optional", "washrooms total override"
    ],
    "washrooms_elec_heat": [
        "how many washrooms have electric heating", "washrooms with electric heating"
    ],
    "single_door_fridges": [
        "how many single door fridges", "single door fridges"
    ],
    "double_door_fridges": [
        "how many double door fridges", "double door fridges"
    ],
    "small_deep_freezers": [
        "how many small deep freezers", "small deep freezers"
    ],
    "medium_deep_freezers": [
        "how many medium deep freezers", "medium deep freezers"
    ],
    "large_deep_freezers": [
        "how many large deep freezers", "large deep freezers"
    ],
    "primary_cooling": [
        "primary source of cooling", "source of cooling"
    ],
    "ceiling_fans": [
        "how many ceiling fans", "ceiling fans"
    ],
    "standing_fans": [
        "how many standing fans", "standing fans"
    ],
    "fans_on_5pm_7am": [
        "how many fans are usually on from 5pm to 7am", "fans usually on from 5pm to 7am"
    ],
    "ac_1hp": [
        "how many 1 hp acs", "1 hp acs"
    ],
    "ac_1_5hp": [
        "how many 1 5 hp acs", "how many 1.5 hp acs", "1.5 hp acs"
    ],
    "ac_2hp": [
        "how many 2 hp acs", "2 hp acs"
    ],
    "ac_2_5hp": [
        "how many 2 5 hp acs", "how many 2.5 hp acs", "2.5 hp acs"
    ],
    "ac_3hp": [
        "how many 3 hp acs", "3 hp acs"
    ],
    "acs_on_evening": [
        "how many acs are usually on from 5pm to 10 30pm",
        "how many acs are usually on from 5pm to 10:30pm"
    ],
    "acs_on_night": [
        "how many acs are usually on from 10 30pm to 8am",
        "how many acs are usually on from 10:30pm to 8am"
    ],
    "ac_type_response": [
        "what type of acs do you have", "type of acs"
    ],
    "mixed_inverter_acs": [
        "if mixed how many are inverter acs"
    ],
    "mixed_fixed_acs": [
        "if mixed how many are fixed speed acs", "if mixed how many are fixed-speed acs"
    ],
    "ac_temperature_setting": [
        "ac temperature setting optional", "ac temperature setting"
    ],
    "precool_reduction_factor": [
        "pre cooling reduction factor optional", "pre-cooling reduction factor optional", "pre cooling reduction factor"
    ],
    "include_pumps": [
        "do you want to include water pumps"
    ],
    "small_pumps": [
        "how many small water pumps"
    ],
    "medium_pumps": [
        "how many medium water pumps"
    ],
    "large_pumps": [
        "how many large water pumps"
    ],
    "pump_runs_peak": [
        "does any water pump run during peak hours"
    ],
    "pump_runs_outage": [
        "does any water pump run during outages"
    ],
    "washing_machines": [
        "how many washing machines"
    ],
    "irons": [
        "how many irons"
    ],
    "electric_water_heaters": [
        "how many electric water heaters"
    ],
    "microwaves": [
        "how many microwaves"
    ],
    "electric_kettles": [
        "how many electric kettles"
    ],
    "coffee_machines": [
        "how many coffee machines"
    ],
    "blenders": [
        "how many blenders"
    ],
    "juicers": [
        "how many juicers"
    ],
    "air_fryers": [
        "how many air fryers"
    ],
    "dishwashers": [
        "how many dishwashers"
    ],
    "electric_cookers": [
        "how many cookers hot plates", "how many cookers / hot plates"
    ],
    "microwave_evening": [
        "microwave used in the evening"
    ],
    "electric_kettle_evening": [
        "electric kettle used in the evening"
    ],
    "coffee_machine_evening": [
        "coffee machine used in the evening"
    ],
    "blender_evening": [
        "blender used in the evening"
    ],
    "juicer_evening": [
        "juicer used in the evening"
    ],
    "air_fryer_evening": [
        "air fryer used in the evening"
    ],
    "dishwasher_evening": [
        "dishwasher used in the evening"
    ],
    "electric_cooker_evening": [
        "cooker hot plate used in the evening", "cooker / hot plate used in the evening"
    ],
    "iron_evening": [
        "iron used in the evening"
    ],
    "washing_machine_evening": [
        "washing machine used in the evening"
    ],
    "water_heater_evening": [
        "water heater used in the evening"
    ],
    "water_pump_evening": [
        "water pump used in the evening"
    ],
}

def resolve_column(alias_key: str):
    aliases = COLUMN_ALIASES.get(alias_key, [])
    for alias in aliases:
        norm_alias = norm_text(alias)
        if norm_alias in normalized_column_index:
            return normalized_column_index[norm_alias]
    return None

RESOLVED_COLUMNS = {k: resolve_column(k) for k in COLUMN_ALIASES.keys()}

def get_num(alias_key: str, default=0.0):
    col = RESOLVED_COLUMNS.get(alias_key)
    if col is None:
        return default
    return n(row.get(col), default)

def get_text(alias_key: str, default=""):
    col = RESOLVED_COLUMNS.get(alias_key)
    if col is None:
        return default
    return s(row.get(col))

column_audit_path = os.path.join(load_dir, "column_mapping_audit.json")
save_json(column_audit_path, {
    "available_columns": available_columns,
    "resolved_columns": RESOLVED_COLUMNS,
})

# =========================
# BLOCK 2B: CORE LOAD / SURGE / BATTERY ENGINE
# =========================

POWER_KW = {
    "ac_1hp": 0.90,
    "ac_1_5hp": 1.20,
    "ac_2hp": 1.60,
    "ac_2_5hp": 2.00,
    "ac_3hp": 2.40,
    "ceiling_fan": 0.06,
    "standing_fan": 0.08,
    "single_door_fridge": 0.15,
    "double_door_fridge": 0.20,
    "small_deep_freezer": 0.15,
    "medium_deep_freezer": 0.20,
    "large_deep_freezer": 0.30,
    "router": 0.015,
    "small_device_charging_bundle": 0.02,
    "tv": 0.10,
    "hall_lighting": 0.06,
    "bedroom_lighting": 0.02,
    "kitchen_lighting": 0.02,
    "washroom_lighting": 0.015,
    "outdoor_lighting": 0.05,
    "microwave": 1.2,
    "electric_kettle": 2.0,
    "coffee_machine": 1.0,
    "blender": 0.40,
    "juicer": 0.50,
    "air_fryer": 1.5,
    "dishwasher": 1.5,
    "electric_cooker": 2.0,
    "iron": 1.2,
    "washing_machine": 0.8,
    "electric_water_heater": 2.0,
    "small_pump": 0.75,
    "medium_pump": 1.10,
    "large_pump": 1.50,
}

HOURS = {
    "evening_period": 5.5,
    "night_period": 9.5,
    "full_window": 15.0,
    "router": 15.0,
    "charging_bundle": 3.0,
    "hall_lighting": 5.0,
    "bedroom_lighting": 3.0,
    "kitchen_lighting": 2.0,
    "washroom_lighting": 1.0,
    "outdoor_lighting": 12.0,
    "tv_per_hall": 3.0,   # 8pm to 11pm
    "microwave": 0.25,
    "electric_kettle": 0.25,
    "coffee_machine": 0.10,
    "blender": 0.10,
    "juicer": 0.10,
    "air_fryer": 0.50,
    "dishwasher": 1.50,
    "electric_cooker": 1.50,
    "iron": 0.50,
    "washing_machine": 1.00,
    "electric_water_heater": 1.00,
    "small_pump": 0.50,
    "medium_pump": 0.50,
    "large_pump": 0.50,
}

STARTUP_MULTIPLIERS = {
    "inverter_ac": 1.5,
    "fixed_speed_ac": 3.5,
    "unknown_ac": 2.5,
    "fridge_freezer": 2.0,
    "pump": 3.0,
    "washing_machine": 1.8,
    "fan": 1.2,
}

MARGINS = {
    "inv_minimum": 1.10,
    "inv_balanced": 1.20,
    "inv_future": 1.35,
    "bat_minimum": 1.05,
    "bat_balanced": 1.15,
    "bat_future": 1.30,
}

BATTERY_EFFICIENCY = 0.90
BATTERY_USABLE_DOD = 0.90
INVERTER_DELIVERY_EFFICIENCY = 0.94
SINGLE_PHASE_VOLTAGE = 230.0
INVERTER_CLASSES_SINGLE_PHASE = [3.5, 5.0, 6.5, 8.0, 10.0]
INVERTER_CLASSES_THREE_PHASE = [8.0, 10.0, 12.0, 15.0, 20.0]
DEFAULT_TEMP_SETTING = 26.0
REFRIGERATION_CYCLING_FACTOR = 0.50
FAN_ENERGY_FACTOR = 0.85
COOLING_FACTORS = {
    "inverter_ac": {"evening": 0.42, "night": 0.32},
    "fixed_speed_ac": {"evening": 0.62, "night": 0.50},
    "unknown_ac": {"evening": 0.52, "night": 0.40},
}
DEFAULT_PRECOOL_REDUCTION_FACTOR = 0.00

def pick_inverter_class(required_kw: float, supply_phase: str):
    classes = INVERTER_CLASSES_SINGLE_PHASE if "single" in supply_phase.lower() else INVERTER_CLASSES_THREE_PHASE
    for size in classes:
        if size >= required_kw:
            return size
    return None

def largest_selected_event_power(event_rows: Dict[str, float]) -> float:
    return max(event_rows.values()) if event_rows else 0.0

def battery_nominal_from_required_energy(e_required_kwh_ac: float) -> float:
    denom = INVERTER_DELIVERY_EFFICIENCY * BATTERY_EFFICIENCY * BATTERY_USABLE_DOD
    return e_required_kwh_ac / denom if denom > 0 else 0.0

def extract_submission_fields() -> Dict[str, Any]:
    bedrooms = get_num("bedrooms", 0.0)
    washrooms_override = get_num("washrooms_override", 0.0)
    washrooms = washrooms_override or max(1.0, bedrooms)
    return {
        "site_name": get_text("site_name"),
        "customer_name": get_text("customer_name"),
        "supply_phase": get_text("supply_phase"),
        "system_priority": get_text("system_priority"),
        "bedrooms": bedrooms,
        "halls": get_num("halls", 0.0),
        "kitchens": get_num("kitchens", 0.0),
        "washrooms": washrooms,
        "washrooms_elec_heat": get_num("washrooms_elec_heat", 0.0),
        "single_door_fridges": get_num("single_door_fridges", 0.0),
        "double_door_fridges": get_num("double_door_fridges", 0.0),
        "small_deep_freezers": get_num("small_deep_freezers", 0.0),
        "medium_deep_freezers": get_num("medium_deep_freezers", 0.0),
        "large_deep_freezers": get_num("large_deep_freezers", 0.0),
        "primary_cooling": get_text("primary_cooling"),
        "ceiling_fans": get_num("ceiling_fans", 0.0),
        "standing_fans": get_num("standing_fans", 0.0),
        "fans_on_5pm_7am": get_num("fans_on_5pm_7am", 0.0),
        "ac_1hp": get_num("ac_1hp", 0.0),
        "ac_1_5hp": get_num("ac_1_5hp", 0.0),
        "ac_2hp": get_num("ac_2hp", 0.0),
        "ac_2_5hp": get_num("ac_2_5hp", 0.0),
        "ac_3hp": get_num("ac_3hp", 0.0),
        "acs_on_evening": get_num("acs_on_evening", 0.0),
        "acs_on_night": get_num("acs_on_night", 0.0),
        "ac_type_response": get_text("ac_type_response"),
        "mixed_inverter_acs": get_num("mixed_inverter_acs", 0.0),
        "mixed_fixed_acs": get_num("mixed_fixed_acs", 0.0),
        "ac_temperature_setting": get_num("ac_temperature_setting", DEFAULT_TEMP_SETTING),
        "precool_reduction_factor": get_num("precool_reduction_factor", DEFAULT_PRECOOL_REDUCTION_FACTOR),
        "include_pumps": get_text("include_pumps"),
        "small_pumps": get_num("small_pumps", 0.0),
        "medium_pumps": get_num("medium_pumps", 0.0),
        "large_pumps": get_num("large_pumps", 0.0),
        "pump_runs_peak": get_text("pump_runs_peak"),
        "pump_runs_outage": get_text("pump_runs_outage"),
        "washing_machines": get_num("washing_machines", 0.0),
        "irons": get_num("irons", 0.0),
        "electric_water_heaters": get_num("electric_water_heaters", 0.0),
        "microwaves": get_num("microwaves", 0.0),
        "electric_kettles": get_num("electric_kettles", 0.0),
        "coffee_machines": get_num("coffee_machines", 0.0),
        "blenders": get_num("blenders", 0.0),
        "juicers": get_num("juicers", 0.0),
        "air_fryers": get_num("air_fryers", 0.0),
        "dishwashers": get_num("dishwashers", 0.0),
        "electric_cookers": get_num("electric_cookers", 0.0),
        "microwave_evening": get_num("microwave_evening", 0.0),
        "electric_kettle_evening": get_num("electric_kettle_evening", 0.0),
        "coffee_machine_evening": get_num("coffee_machine_evening", 0.0),
        "blender_evening": get_num("blender_evening", 0.0),
        "juicer_evening": get_num("juicer_evening", 0.0),
        "air_fryer_evening": get_num("air_fryer_evening", 0.0),
        "dishwasher_evening": get_num("dishwasher_evening", 0.0),
        "electric_cooker_evening": get_num("electric_cooker_evening", 0.0),
        "iron_evening": get_num("iron_evening", 0.0),
        "washing_machine_evening": get_num("washing_machine_evening", 0.0),
        "water_heater_evening": get_num("water_heater_evening", 0.0),
        "water_pump_evening": get_num("water_pump_evening", 0.0),
    }

def infer_ac_type_category(fields: Dict[str, Any]) -> str:
    txt = fields["ac_type_response"].lower()
    if "all inverter" in txt:
        return "inverter_ac"
    if "all fixed" in txt:
        return "fixed_speed_ac"
    return "unknown_ac"

def ac_type_mix_counts(fields: Dict[str, Any]) -> Dict[str, float]:
    total_acs = fields["ac_1hp"] + fields["ac_1_5hp"] + fields["ac_2hp"] + fields["ac_2_5hp"] + fields["ac_3hp"]
    category = infer_ac_type_category(fields)
    if total_acs <= 0:
        return {"inverter": 0.0, "fixed": 0.0, "unknown": 0.0}
    if category == "inverter_ac":
        return {"inverter": total_acs, "fixed": 0.0, "unknown": 0.0}
    if category == "fixed_speed_ac":
        return {"inverter": 0.0, "fixed": total_acs, "unknown": 0.0}
    mixed_inverter = min(fields["mixed_inverter_acs"], total_acs)
    mixed_fixed = min(fields["mixed_fixed_acs"], max(0.0, total_acs - mixed_inverter))
    unknown = max(0.0, total_acs - mixed_inverter - mixed_fixed)
    return {"inverter": mixed_inverter, "fixed": mixed_fixed, "unknown": unknown}

def recruited_refrigeration(fields: Dict[str, Any]) -> Dict[str, float]:
    return {
        "single_door_fridges": fields["single_door_fridges"] * POWER_KW["single_door_fridge"],
        "double_door_fridges": fields["double_door_fridges"] * POWER_KW["double_door_fridge"],
        "small_deep_freezers": fields["small_deep_freezers"] * POWER_KW["small_deep_freezer"],
        "medium_deep_freezers": fields["medium_deep_freezers"] * POWER_KW["medium_deep_freezer"],
        "large_deep_freezers": fields["large_deep_freezers"] * POWER_KW["large_deep_freezer"],
    }

def inferred_lighting(fields: Dict[str, Any]) -> Dict[str, float]:
    return {
        "hall_lighting": fields["halls"] * POWER_KW["hall_lighting"],
        "bedroom_lighting": fields["bedrooms"] * POWER_KW["bedroom_lighting"],
        "kitchen_lighting": fields["kitchens"] * POWER_KW["kitchen_lighting"],
        "washroom_lighting": fields["washrooms"] * POWER_KW["washroom_lighting"],
        "outdoor_lighting": POWER_KW["outdoor_lighting"],
    }

def auto_always_on(fields: Dict[str, Any]) -> Dict[str, float]:
    return {
        "router": POWER_KW["router"],
        "charging_bundle": POWER_KW["small_device_charging_bundle"],
        "tv": fields["halls"] * POWER_KW["tv"],
    }

def ac_inventory_power(fields: Dict[str, Any]) -> List[Tuple[str, float]]:
    inventory = []
    for key, power_key in [
        ("ac_1hp", "ac_1hp"),
        ("ac_1_5hp", "ac_1_5hp"),
        ("ac_2hp", "ac_2hp"),
        ("ac_2_5hp", "ac_2_5hp"),
        ("ac_3hp", "ac_3hp"),
    ]:
        count = int(round(fields[key]))
        for _ in range(count):
            inventory.append((key, POWER_KW[power_key]))
    return inventory

def active_ac_power(fields: Dict[str, Any]) -> Tuple[float, float]:
    inventory = sorted([p for _, p in ac_inventory_power(fields)], reverse=True)
    evening_count = int(round(fields["acs_on_evening"]))
    night_count = int(round(fields["acs_on_night"]))
    evening_power = sum(inventory[:evening_count]) if evening_count > 0 else 0.0
    night_power = sum(inventory[:night_count]) if night_count > 0 else 0.0
    return evening_power, night_power

def active_fan_power(fields: Dict[str, Any]) -> float:
    active_fans = int(round(fields["fans_on_5pm_7am"]))
    return active_fans * POWER_KW["ceiling_fan"]

def cooling_energy_from_ac(fields: Dict[str, Any], evening_power_kw: float, night_power_kw: float) -> Dict[str, float]:
    ac_mix = ac_type_mix_counts(fields)
    total_mix = ac_mix["inverter"] + ac_mix["fixed"] + ac_mix["unknown"]
    if total_mix <= 0:
        return {
            "cooling_energy_evening_kwh": 0.0,
            "cooling_energy_night_kwh": 0.0,
            "cooling_energy_kwh": 0.0,
            "cooling_evening_factor_effective": 0.0,
            "cooling_night_factor_effective": 0.0,
            "ac_inverter_share": 0.0,
            "ac_fixed_share": 0.0,
            "ac_unknown_share": 0.0,
            "temp_setting_factor": 1.0,
            "precool_reduction_factor_used": 0.0,
        }

    inverter_share = ac_mix["inverter"] / total_mix
    fixed_share = ac_mix["fixed"] / total_mix
    unknown_share = ac_mix["unknown"] / total_mix
    evening_factor = inverter_share * COOLING_FACTORS["inverter_ac"]["evening"] + fixed_share * COOLING_FACTORS["fixed_speed_ac"]["evening"] + unknown_share * COOLING_FACTORS["unknown_ac"]["evening"]
    night_factor = inverter_share * COOLING_FACTORS["inverter_ac"]["night"] + fixed_share * COOLING_FACTORS["fixed_speed_ac"]["night"] + unknown_share * COOLING_FACTORS["unknown_ac"]["night"]

    temperature_setting = n(fields.get("ac_temperature_setting"), DEFAULT_TEMP_SETTING)
    temp_setting_factor = 26.0 / temperature_setting if temperature_setting > 0 else 1.0
    precool_reduction = max(0.0, min(0.6, n(fields.get("precool_reduction_factor"), DEFAULT_PRECOOL_REDUCTION_FACTOR)))

    evening_energy = evening_power_kw * HOURS["evening_period"] * evening_factor * temp_setting_factor
    night_energy = night_power_kw * HOURS["night_period"] * night_factor * temp_setting_factor
    evening_energy *= (1.0 - precool_reduction)
    total = evening_energy + night_energy

    return {
        "cooling_energy_evening_kwh": evening_energy,
        "cooling_energy_night_kwh": night_energy,
        "cooling_energy_kwh": total,
        "cooling_evening_factor_effective": evening_factor,
        "cooling_night_factor_effective": night_factor,
        "ac_inverter_share": inverter_share,
        "ac_fixed_share": fixed_share,
        "ac_unknown_share": unknown_share,
        "temp_setting_factor": temp_setting_factor,
        "precool_reduction_factor_used": precool_reduction,
    }

def selected_event_loads(fields: Dict[str, Any]) -> Dict[str, float]:
    events = {}
    def add_if_selected(flag_key: str, qty_key: str, power_key: str, name: str):
        if fields.get(flag_key, 0.0) > 0 and fields.get(qty_key, 0.0) > 0:
            events[name] = fields[qty_key] * POWER_KW[power_key]

    add_if_selected("microwave_evening", "microwaves", "microwave", "microwave")
    add_if_selected("electric_kettle_evening", "electric_kettles", "electric_kettle", "electric_kettle")
    add_if_selected("coffee_machine_evening", "coffee_machines", "coffee_machine", "coffee_machine")
    add_if_selected("blender_evening", "blenders", "blender", "blender")
    add_if_selected("juicer_evening", "juicers", "juicer", "juicer")
    add_if_selected("air_fryer_evening", "air_fryers", "air_fryer", "air_fryer")
    add_if_selected("dishwasher_evening", "dishwashers", "dishwasher", "dishwasher")
    add_if_selected("electric_cooker_evening", "electric_cookers", "electric_cooker", "electric_cooker")
    add_if_selected("iron_evening", "irons", "iron", "iron")
    add_if_selected("washing_machine_evening", "washing_machines", "washing_machine", "washing_machine")
    add_if_selected("water_heater_evening", "electric_water_heaters", "electric_water_heater", "electric_water_heater")

    if fields.get("water_pump_evening", 0.0) > 0 and "yes" in fields.get("include_pumps", "").lower():
        if fields["large_pumps"] > 0:
            events["large_pump"] = fields["large_pumps"] * POWER_KW["large_pump"]
        elif fields["medium_pumps"] > 0:
            events["medium_pump"] = fields["medium_pumps"] * POWER_KW["medium_pump"]
        elif fields["small_pumps"] > 0:
            events["small_pump"] = fields["small_pumps"] * POWER_KW["small_pump"]
    return events

def calc_baseline_power(fields: Dict[str, Any]) -> Tuple[float, Dict[str, float]]:
    refr = recruited_refrigeration(fields)
    lights = inferred_lighting(fields)
    auto = auto_always_on(fields)
    detail = {}
    detail.update(refr)
    detail.update(lights)
    detail.update(auto)
    return sum(detail.values()), detail

def calc_running_and_peak(fields: Dict[str, Any]) -> Dict[str, Any]:
    baseline_kw, baseline_detail = calc_baseline_power(fields)
    if "ac" in fields["primary_cooling"].lower():
        cooling_evening_kw, cooling_night_kw = active_ac_power(fields)
        cooling_peak_kw = max(cooling_evening_kw, cooling_night_kw)
        cooling_energy_detail = cooling_energy_from_ac(fields, cooling_evening_kw, cooling_night_kw)
        cooling_energy_kwh = cooling_energy_detail["cooling_energy_kwh"]
        cooling_label = "AC"
    else:
        cooling_peak_kw = active_fan_power(fields)
        cooling_evening_kw = cooling_peak_kw
        cooling_night_kw = cooling_peak_kw
        cooling_energy_kwh = cooling_peak_kw * HOURS["full_window"] * FAN_ENERGY_FACTOR
        cooling_energy_detail = {
            "cooling_energy_evening_kwh": cooling_peak_kw * HOURS["evening_period"] * FAN_ENERGY_FACTOR,
            "cooling_energy_night_kwh": cooling_peak_kw * HOURS["night_period"] * FAN_ENERGY_FACTOR,
            "cooling_energy_kwh": cooling_energy_kwh,
            "cooling_evening_factor_effective": FAN_ENERGY_FACTOR,
            "cooling_night_factor_effective": FAN_ENERGY_FACTOR,
            "ac_inverter_share": 0.0,
            "ac_fixed_share": 0.0,
            "ac_unknown_share": 0.0,
            "temp_setting_factor": 1.0,
            "precool_reduction_factor_used": 0.0,
        }
        cooling_label = "Fans"

    event_detail = selected_event_loads(fields)
    event_peak_kw = largest_selected_event_power(event_detail)
    p_running = baseline_kw + cooling_peak_kw
    p_peak_run = p_running + event_peak_kw

    return {
        "baseline_kw": baseline_kw,
        "cooling_evening_kw": cooling_evening_kw,
        "cooling_night_kw": cooling_night_kw,
        "cooling_peak_kw": cooling_peak_kw,
        "cooling_energy_kwh": cooling_energy_kwh,
        "p_running_kw": p_running,
        "p_peak_run_kw": p_peak_run,
        "event_peak_kw": event_peak_kw,
        "cooling_mode": cooling_label,
        "baseline_detail": baseline_detail,
        "event_detail": event_detail,
        **cooling_energy_detail,
    }

def calc_surge(fields: Dict[str, Any], p_peak_run_kw: float) -> Dict[str, float]:
    ac_type_cat = infer_ac_type_category(fields)
    ac_multiplier = STARTUP_MULTIPLIERS[ac_type_cat]
    ac_units = sorted([p for _, p in ac_inventory_power(fields)], reverse=True)
    largest_ac_kw = ac_units[0] if ac_units else 0.0

    refr_units = []
    refr_units += [POWER_KW["single_door_fridge"]] * int(round(fields["single_door_fridges"]))
    refr_units += [POWER_KW["double_door_fridge"]] * int(round(fields["double_door_fridges"]))
    refr_units += [POWER_KW["small_deep_freezer"]] * int(round(fields["small_deep_freezers"]))
    refr_units += [POWER_KW["medium_deep_freezer"]] * int(round(fields["medium_deep_freezers"]))
    refr_units += [POWER_KW["large_deep_freezer"]] * int(round(fields["large_deep_freezers"]))
    largest_refr_kw = max(refr_units) if refr_units else 0.0

    pump_units = []
    pump_units += [POWER_KW["small_pump"]] * int(round(fields["small_pumps"]))
    pump_units += [POWER_KW["medium_pump"]] * int(round(fields["medium_pumps"]))
    pump_units += [POWER_KW["large_pump"]] * int(round(fields["large_pumps"]))
    largest_pump_kw = max(pump_units) if pump_units else 0.0

    wm_kw = POWER_KW["washing_machine"] if fields["washing_machines"] > 0 else 0.0

    ac_extra = largest_ac_kw * (ac_multiplier - 1) if largest_ac_kw > 0 else 0.0
    refr_extra = largest_refr_kw * (STARTUP_MULTIPLIERS["fridge_freezer"] - 1) if largest_refr_kw > 0 else 0.0
    pump_extra = largest_pump_kw * (STARTUP_MULTIPLIERS["pump"] - 1) if largest_pump_kw > 0 else 0.0
    wm_extra = wm_kw * (STARTUP_MULTIPLIERS["washing_machine"] - 1) if wm_kw > 0 else 0.0

    s1 = p_peak_run_kw + ac_extra
    s2 = p_peak_run_kw + ac_extra + refr_extra
    s3 = p_peak_run_kw + pump_extra
    s4 = p_peak_run_kw + pump_extra + refr_extra
    s5 = p_peak_run_kw + wm_extra

    p_surge_design = max(s1, s2, s3, s4, s5)
    i_surge_design = (p_surge_design * 1000.0) / SINGLE_PHASE_VOLTAGE if "single" in fields["supply_phase"].lower() else float("nan")

    return {
        "largest_ac_kw": largest_ac_kw,
        "largest_refr_kw": largest_refr_kw,
        "largest_pump_kw": largest_pump_kw,
        "ac_startup_multiplier": ac_multiplier,
        "s1_kw": s1,
        "s2_kw": s2,
        "s3_kw": s3,
        "s4_kw": s4,
        "s5_kw": s5,
        "p_surge_design_kw": p_surge_design,
        "i_surge_design_a": i_surge_design,
    }

def calc_energy_and_battery(fields: Dict[str, Any], running_peak: Dict[str, Any]) -> Dict[str, float]:
    baseline_detail = running_peak["baseline_detail"]
    event_detail = running_peak["event_detail"]

    e_refrigeration = (
        baseline_detail.get("single_door_fridges", 0.0) * HOURS["full_window"] +
        baseline_detail.get("double_door_fridges", 0.0) * HOURS["full_window"] +
        baseline_detail.get("small_deep_freezers", 0.0) * HOURS["full_window"] +
        baseline_detail.get("medium_deep_freezers", 0.0) * HOURS["full_window"] +
        baseline_detail.get("large_deep_freezers", 0.0) * HOURS["full_window"]
    ) * REFRIGERATION_CYCLING_FACTOR

    e_router = baseline_detail.get("router", 0.0) * HOURS["router"]
    e_charging = baseline_detail.get("charging_bundle", 0.0) * HOURS["charging_bundle"]
    e_lighting = (
        baseline_detail.get("hall_lighting", 0.0) * HOURS["hall_lighting"] +
        baseline_detail.get("bedroom_lighting", 0.0) * HOURS["bedroom_lighting"] +
        baseline_detail.get("kitchen_lighting", 0.0) * HOURS["kitchen_lighting"] +
        baseline_detail.get("washroom_lighting", 0.0) * HOURS["washroom_lighting"] +
        baseline_detail.get("outdoor_lighting", 0.0) * HOURS["outdoor_lighting"]
    )
    e_tv = baseline_detail.get("tv", 0.0) * HOURS["tv_per_hall"]
    e_cooling = running_peak["cooling_energy_kwh"]

    duration_lookup = {
        "microwave": HOURS["microwave"],
        "electric_kettle": HOURS["electric_kettle"],
        "coffee_machine": HOURS["coffee_machine"],
        "blender": HOURS["blender"],
        "juicer": HOURS["juicer"],
        "air_fryer": HOURS["air_fryer"],
        "dishwasher": HOURS["dishwasher"],
        "electric_cooker": HOURS["electric_cooker"],
        "iron": HOURS["iron"],
        "washing_machine": HOURS["washing_machine"],
        "electric_water_heater": HOURS["electric_water_heater"],
        "small_pump": HOURS["small_pump"],
        "medium_pump": HOURS["medium_pump"],
        "large_pump": HOURS["large_pump"],
    }

    e_events = 0.0
    for name, kw in event_detail.items():
        e_events += kw * duration_lookup.get(name, 0.0)

    e_required_kwh = e_refrigeration + e_router + e_charging + e_lighting + e_tv + e_cooling + e_events
    battery_nominal = battery_nominal_from_required_energy(e_required_kwh)

    return {
        "e_refrigeration_kwh": e_refrigeration,
        "e_router_kwh": e_router,
        "e_charging_kwh": e_charging,
        "e_lighting_kwh": e_lighting,
        "e_tv_kwh": e_tv,
        "e_cooling_kwh": e_cooling,
        "e_events_kwh": e_events,
        "e_required_kwh": e_required_kwh,
        "battery_nominal_kwh": battery_nominal,
        "battery_minimum_kwh": battery_nominal * MARGINS["bat_minimum"],
        "battery_balanced_kwh": battery_nominal * MARGINS["bat_balanced"],
        "battery_future_kwh": battery_nominal * MARGINS["bat_future"],
        "inverter_delivery_efficiency_used": INVERTER_DELIVERY_EFFICIENCY,
        "battery_efficiency_used": BATTERY_EFFICIENCY,
        "battery_usable_dod_used": BATTERY_USABLE_DOD,
    }

def calc_recommendations(fields: Dict[str, Any], p_peak_run_kw: float, p_surge_design_kw: float) -> Dict[str, Any]:
    p_inv_minimal = p_peak_run_kw * MARGINS["inv_minimum"]
    p_inv_balanced = p_peak_run_kw * MARGINS["inv_balanced"]
    p_inv_future = p_peak_run_kw * MARGINS["inv_future"]
    rec = {}

    for label, required in [("minimum", p_inv_minimal), ("balanced", p_inv_balanced), ("future", p_inv_future)]:
        klass = pick_inverter_class(required, fields["supply_phase"])
        if klass is not None:
            classes = INVERTER_CLASSES_SINGLE_PHASE if "single" in fields["supply_phase"].lower() else INVERTER_CLASSES_THREE_PHASE
            idx = classes.index(klass)
            while idx < len(classes) and classes[idx] < max(required, p_surge_design_kw):
                idx += 1
            klass = classes[idx] if idx < len(classes) else None

        rec[f"p_inv_{label}_required_kw"] = required
        rec[f"inverter_{label}_kw"] = klass
    return rec

fields = extract_submission_fields()
running_peak = calc_running_and_peak(fields)
surge = calc_surge(fields, running_peak["p_peak_run_kw"])
battery = calc_energy_and_battery(fields, running_peak)
recs = calc_recommendations(fields, running_peak["p_peak_run_kw"], surge["p_surge_design_kw"])

result = {}
result.update(fields)
result.update({
    "baseline_kw": running_peak["baseline_kw"],
    "cooling_evening_kw": running_peak["cooling_evening_kw"],
    "cooling_night_kw": running_peak["cooling_night_kw"],
    "cooling_peak_kw": running_peak["cooling_peak_kw"],
    "cooling_energy_evening_kwh": running_peak["cooling_energy_evening_kwh"],
    "cooling_energy_night_kwh": running_peak["cooling_energy_night_kwh"],
    "cooling_evening_factor_effective": running_peak["cooling_evening_factor_effective"],
    "cooling_night_factor_effective": running_peak["cooling_night_factor_effective"],
    "ac_inverter_share": running_peak["ac_inverter_share"],
    "ac_fixed_share": running_peak["ac_fixed_share"],
    "ac_unknown_share": running_peak["ac_unknown_share"],
    "temp_setting_factor": running_peak["temp_setting_factor"],
    "precool_reduction_factor_used": running_peak["precool_reduction_factor_used"],
    "p_running_kw": running_peak["p_running_kw"],
    "event_peak_kw": running_peak["event_peak_kw"],
    "p_peak_run_kw": running_peak["p_peak_run_kw"],
    "cooling_mode": running_peak["cooling_mode"],
})
result.update(surge)
result.update(battery)
result.update(recs)

# =========================
# BLOCK 2C: CANONICAL JSON + MD OUTPUTS
# =========================

load_json_path = os.path.join(load_dir, "load_state.json")
load_report_md_path = os.path.join(load_dir, "load_report.md")
load_snapshot_png_path = os.path.join(load_dir, "load_snapshot_3x3.png")

load_state = {
    "meta": {
        "generated_at": datetime.now().isoformat(),
        "submission_key": latest_key,
        "project_root": project_root,
        "source_raw_state": raw_state_path,
        "source_rebuilt_row": rebuilt_kobo_row_path,
        "project_name": result.get("site_name"),
        "customer_name": result.get("customer_name"),
        "column_mapping_audit": column_audit_path,
    },
    "inputs_used": {
        "site_name": result.get("site_name"),
        "customer_name": result.get("customer_name"),
        "supply_phase": result.get("supply_phase"),
        "system_priority": result.get("system_priority"),
        "bedrooms": result.get("bedrooms"),
        "halls": result.get("halls"),
        "kitchens": result.get("kitchens"),
        "washrooms": result.get("washrooms"),
        "single_door_fridges": result.get("single_door_fridges"),
        "double_door_fridges": result.get("double_door_fridges"),
        "small_deep_freezers": result.get("small_deep_freezers"),
        "medium_deep_freezers": result.get("medium_deep_freezers"),
        "large_deep_freezers": result.get("large_deep_freezers"),
        "primary_cooling": result.get("primary_cooling"),
        "ceiling_fans": result.get("ceiling_fans"),
        "standing_fans": result.get("standing_fans"),
        "fans_on_5pm_7am": result.get("fans_on_5pm_7am"),
        "ac_1hp": result.get("ac_1hp"),
        "ac_1_5hp": result.get("ac_1_5hp"),
        "ac_2hp": result.get("ac_2hp"),
        "ac_2_5hp": result.get("ac_2_5hp"),
        "ac_3hp": result.get("ac_3hp"),
        "acs_on_evening": result.get("acs_on_evening"),
        "acs_on_night": result.get("acs_on_night"),
        "ac_type_response": result.get("ac_type_response"),
        "mixed_inverter_acs": result.get("mixed_inverter_acs"),
        "mixed_fixed_acs": result.get("mixed_fixed_acs"),
        "ac_temperature_setting": result.get("ac_temperature_setting"),
        "precool_reduction_factor": result.get("precool_reduction_factor"),
    },
    "load_results": {
        "baseline_kw": result.get("baseline_kw"),
        "cooling_evening_kw": result.get("cooling_evening_kw"),
        "cooling_night_kw": result.get("cooling_night_kw"),
        "cooling_peak_kw": result.get("cooling_peak_kw"),
        "cooling_energy_evening_kwh": result.get("cooling_energy_evening_kwh"),
        "cooling_energy_night_kwh": result.get("cooling_energy_night_kwh"),
        "cooling_evening_factor_effective": result.get("cooling_evening_factor_effective"),
        "cooling_night_factor_effective": result.get("cooling_night_factor_effective"),
        "ac_inverter_share": result.get("ac_inverter_share"),
        "ac_fixed_share": result.get("ac_fixed_share"),
        "ac_unknown_share": result.get("ac_unknown_share"),
        "temp_setting_factor": result.get("temp_setting_factor"),
        "precool_reduction_factor_used": result.get("precool_reduction_factor_used"),
        "p_running_kw": result.get("p_running_kw"),
        "event_peak_kw": result.get("event_peak_kw"),
        "p_peak_run_kw": result.get("p_peak_run_kw"),
        "p_surge_design_kw": result.get("p_surge_design_kw"),
        "i_surge_design_a": result.get("i_surge_design_a"),
    },
    "energy_breakdown": {
        "e_refrigeration_kwh": result.get("e_refrigeration_kwh"),
        "e_router_kwh": result.get("e_router_kwh"),
        "e_charging_kwh": result.get("e_charging_kwh"),
        "e_lighting_kwh": result.get("e_lighting_kwh"),
        "e_tv_kwh": result.get("e_tv_kwh"),
        "e_cooling_kwh": result.get("e_cooling_kwh"),
        "e_events_kwh": result.get("e_events_kwh"),
        "e_required_kwh": result.get("e_required_kwh"),
    },
    "battery_outputs": {
        "battery_nominal_kwh": result.get("battery_nominal_kwh"),
        "battery_minimum_kwh": result.get("battery_minimum_kwh"),
        "battery_balanced_kwh": result.get("battery_balanced_kwh"),
        "battery_future_kwh": result.get("battery_future_kwh"),
        "inverter_delivery_efficiency_used": result.get("inverter_delivery_efficiency_used"),
        "battery_efficiency_used": result.get("battery_efficiency_used"),
        "battery_usable_dod_used": result.get("battery_usable_dod_used"),
    },
    "inverter_recommendations": {
        "p_inv_minimum_required_kw": result.get("p_inv_minimum_required_kw"),
        "p_inv_balanced_required_kw": result.get("p_inv_balanced_required_kw"),
        "p_inv_future_required_kw": result.get("p_inv_future_required_kw"),
        "inverter_minimum_kw": result.get("inverter_minimum_kw"),
        "inverter_balanced_kw": result.get("inverter_balanced_kw"),
        "inverter_future_kw": result.get("inverter_future_kw"),
    },
    "artifacts": {
        "load_report_md": load_report_md_path,
        "load_snapshot_png": load_snapshot_png_path,
    }
}

save_json(load_json_path, load_state)

load_report_md = f"""# TENET Load Intelligence Report

## Project
- Project Name: {result.get("site_name")}
- Customer Name: {result.get("customer_name")}
- Supply Phase: {result.get("supply_phase")}
- System Priority: {result.get("system_priority")}

## Core Results
- Peak Running Load: **{result.get("p_running_kw", 0):.2f} kW**
- Peak with Event Load: **{result.get("p_peak_run_kw", 0):.2f} kW**
- Surge Design Load: **{result.get("p_surge_design_kw", 0):.2f} kW**
- Surge Design Current: **{result.get("i_surge_design_a", 0):.2f} A**

## Cooling Results
- Cooling Mode: **{result.get("cooling_mode", "")}**
- Evening Cooling Energy: **{result.get("cooling_energy_evening_kwh", 0):.2f} kWh/day**
- Night Cooling Energy: **{result.get("cooling_energy_night_kwh", 0):.2f} kWh/day**
- Total Cooling Energy: **{result.get("e_cooling_kwh", 0):.2f} kWh/day**
- Inverter AC Share: **{result.get("ac_inverter_share", 0):.2f}**
- Fixed-speed AC Share: **{result.get("ac_fixed_share", 0):.2f}**
- Temperature Factor: **{result.get("temp_setting_factor", 0):.2f}**
- Pre-cool Reduction Used: **{result.get("precool_reduction_factor_used", 0):.2f}**

## Energy Breakdown
- Refrigeration: **{result.get("e_refrigeration_kwh", 0):.2f} kWh/day**
- Router: **{result.get("e_router_kwh", 0):.2f} kWh/day**
- Charging Bundle: **{result.get("e_charging_kwh", 0):.2f} kWh/day**
- Lighting: **{result.get("e_lighting_kwh", 0):.2f} kWh/day**
- TV: **{result.get("e_tv_kwh", 0):.2f} kWh/day**
- Event Loads: **{result.get("e_events_kwh", 0):.2f} kWh/day**
- Total Required AC Energy: **{result.get("e_required_kwh", 0):.2f} kWh/day**

## Battery Outputs
- Nominal Battery Energy: **{result.get("battery_nominal_kwh", 0):.2f} kWh**
- Minimum Battery Energy: **{result.get("battery_minimum_kwh", 0):.2f} kWh**
- Balanced Battery Energy: **{result.get("battery_balanced_kwh", 0):.2f} kWh**
- Future Battery Energy: **{result.get("battery_future_kwh", 0):.2f} kWh**

## Inverter Recommendation
- Minimum Inverter Class: **{result.get("inverter_minimum_kw")} kW**
- Balanced Inverter Class: **{result.get("inverter_balanced_kw")} kW**
- Future Inverter Class: **{result.get("inverter_future_kw")} kW**

## Canonical Handoff
- JSON State: `{load_json_path}`
"""

with open(load_report_md_path, "w", encoding="utf-8") as f:
    f.write(load_report_md)

# =========================
# BLOCK 2D: REFINED 3x3 SNAPSHOT PNG + PRINT
# =========================

hourly = [0.0] * 24

refrigeration_kw_avg = (
    result.get("single_door_fridges", 0) * POWER_KW["single_door_fridge"] +
    result.get("double_door_fridges", 0) * POWER_KW["double_door_fridge"] +
    result.get("small_deep_freezers", 0) * POWER_KW["small_deep_freezer"] +
    result.get("medium_deep_freezers", 0) * POWER_KW["medium_deep_freezer"] +
    result.get("large_deep_freezers", 0) * POWER_KW["large_deep_freezer"]
) * REFRIGERATION_CYCLING_FACTOR

router_kw = POWER_KW["router"]
for h in range(24):
    hourly[h] += refrigeration_kw_avg
    hourly[h] += router_kw

for h in [19, 20, 21]:
    hourly[h] += result.get("e_charging_kwh", 0) / 3.0

hall_light_kw = result.get("halls", 0) * POWER_KW["hall_lighting"]
bedroom_light_kw = result.get("bedrooms", 0) * POWER_KW["bedroom_lighting"]
kitchen_light_kw = result.get("kitchens", 0) * POWER_KW["kitchen_lighting"]
washroom_light_kw = result.get("washrooms", 0) * POWER_KW["washroom_lighting"]
outdoor_light_kw = POWER_KW["outdoor_lighting"]

for h in [18, 19, 20, 21, 22]:
    hourly[h] += hall_light_kw
for h in [20, 21, 22]:
    hourly[h] += bedroom_light_kw
for h in [18, 19]:
    hourly[h] += kitchen_light_kw
for h in [5, 6, 22]:
    hourly[h] += washroom_light_kw / 3.0
for h in [18, 19, 20, 21, 22, 23]:
    hourly[h] += outdoor_light_kw

for h in [20, 21, 22]:
    hourly[h] += result.get("e_tv_kwh", 0) / 3.0

cooling_evening_kw = result.get("cooling_evening_kw", 0)
cooling_night_kw = result.get("cooling_night_kw", 0)
for h in [17, 18, 19, 20, 21, 22]:
    hourly[h] += cooling_evening_kw
for h in [23, 0, 1, 2, 3, 4, 5, 6, 7]:
    hourly[h] += cooling_night_kw

event_hours_map = {
    "microwave": [19],
    "electric_kettle": [19],
    "coffee_machine": [6],
    "blender": [18],
    "juicer": [18],
    "air_fryer": [19, 20],
    "dishwasher": [20, 21],
    "electric_cooker": [18, 19],
    "iron": [20],
    "washing_machine": [19],
    "electric_water_heater": [21],
    "small_pump": [18],
    "medium_pump": [18],
    "large_pump": [18],
}

event_detail = selected_event_loads(fields)
for name, kw in event_detail.items():
    hrs = event_hours_map.get(name, [19])
    share = kw / len(hrs)
    for h in hrs:
        hourly[h] += share

fig, axes = plt.subplots(3, 3, figsize=(17, 12))

axes[0, 0].plot(range(24), hourly, marker="o")
axes[0, 0].set_title("24h Load Profile")
axes[0, 0].set_xlabel("Hour")
axes[0, 0].set_ylabel("kW")
axes[0, 0].set_xticks(range(0, 24, 2))

peak_components = {
    "Baseline": result.get("baseline_kw", 0),
    "Cooling Peak": result.get("cooling_peak_kw", 0),
    "Largest Event": result.get("event_peak_kw", 0),
}
axes[0, 1].bar(list(peak_components.keys()), list(peak_components.values()))
axes[0, 1].set_title("Peak Load Contributors")
axes[0, 1].set_ylabel("kW")
axes[0, 1].tick_params(axis="x", rotation=20)

surge_components = {
    "Largest AC": result.get("largest_ac_kw", 0),
    "Largest Refrig.": result.get("largest_refr_kw", 0),
    "Largest Pump": result.get("largest_pump_kw", 0),
}
axes[0, 2].bar(list(surge_components.keys()), list(surge_components.values()))
axes[0, 2].set_title("Surge Contributors")
axes[0, 2].set_ylabel("kW")
axes[0, 2].tick_params(axis="x", rotation=20)

axes[1, 0].bar(
    ["Refrig.", "Lighting", "TV", "Router", "Charging", "Cooling", "Events"],
    [
        result.get("e_refrigeration_kwh", 0),
        result.get("e_lighting_kwh", 0),
        result.get("e_tv_kwh", 0),
        result.get("e_router_kwh", 0),
        result.get("e_charging_kwh", 0),
        result.get("e_cooling_kwh", 0),
        result.get("e_events_kwh", 0),
    ]
)
axes[1, 0].set_title("Daily Energy Breakdown")
axes[1, 0].set_ylabel("kWh/day")
axes[1, 0].tick_params(axis="x", rotation=20)

axes[1, 1].bar(
    ["Evening Cooling", "Night Cooling"],
    [
        result.get("cooling_energy_evening_kwh", 0),
        result.get("cooling_energy_night_kwh", 0),
    ]
)
axes[1, 1].set_title("Cooling Energy Split")
axes[1, 1].set_ylabel("kWh/day")

axes[1, 2].bar(
    ["1HP", "1.5HP", "2HP", "2.5HP", "3HP"],
    [
        result.get("ac_1hp", 0),
        result.get("ac_1_5hp", 0),
        result.get("ac_2hp", 0),
        result.get("ac_2_5hp", 0),
        result.get("ac_3hp", 0),
    ]
)
axes[1, 2].set_title("AC Inventory")
axes[1, 2].set_ylabel("Units")

axes[2, 0].bar(
    ["Single", "Double", "Small DF", "Medium DF", "Large DF"],
    [
        result.get("single_door_fridges", 0),
        result.get("double_door_fridges", 0),
        result.get("small_deep_freezers", 0),
        result.get("medium_deep_freezers", 0),
        result.get("large_deep_freezers", 0),
    ]
)
axes[2, 0].set_title("Refrigeration Inventory")
axes[2, 0].set_ylabel("Units")
axes[2, 0].tick_params(axis="x", rotation=20)

axes[2, 1].bar(
    ["Batt Nom.", "Batt Bal.", "Inv Min", "Inv Bal", "Inv Fut"],
    [
        result.get("battery_nominal_kwh", 0),
        result.get("battery_balanced_kwh", 0),
        result.get("inverter_minimum_kw", 0) or 0,
        result.get("inverter_balanced_kw", 0) or 0,
        result.get("inverter_future_kw", 0) or 0,
    ]
)
axes[2, 1].set_title("Battery / Inverter")
axes[2, 1].set_ylabel("kWh / kW")
axes[2, 1].tick_params(axis="x", rotation=20)

axes[2, 2].axis("off")
summary_text = (
    f"Project: {result.get('site_name','')}\n"
    f"Customer: {result.get('customer_name','')}\n"
    f"Halls: {result.get('halls',0)}\n\n"
    f"TV Energy:\n{result.get('e_tv_kwh',0):.2f} kWh/day\n\n"
    f"Required Energy:\n{result.get('e_required_kwh',0):.2f} kWh/day\n\n"
    f"Peak+Event:\n{result.get('p_peak_run_kw',0):.2f} kW\n"
    f"Surge:\n{result.get('p_surge_design_kw',0):.2f} kW"
)
axes[2, 2].text(0.03, 0.95, summary_text, va="top", fontsize=11)
axes[2, 2].set_title("Summary")

plt.suptitle("TENET Load Intelligence Snapshot", fontsize=16, y=1.02)
plt.tight_layout()
plt.savefig(load_snapshot_png_path, dpi=180, bbox_inches="tight")
plt.close()

print("----- CANONICAL BLOCK 2 OUTPUTS -----")
print("JSON:", load_json_path)
print("MD:", load_report_md_path)
print("PNG:", load_snapshot_png_path)
print()
print("DEBUG")
print("Resolved hall column:", RESOLVED_COLUMNS.get("halls"))
print("Raw halls value:", row.get(RESOLVED_COLUMNS.get("halls")) if RESOLVED_COLUMNS.get("halls") else None)
print("Extracted halls:", result.get("halls"))
print("TV kWh:", result.get("e_tv_kwh"))
print()
print("Total Required AC Energy (kWh/day):", round(result.get("e_required_kwh", 0), 2))
```

## Block 3A: Inverter Selection

```python
# =========================
# BLOCK 3: INVERTER SELECTION
# FULL REVISED SCRIPT
# Carries full selected inverter schema into JSON handoff
# =========================


import os, json, math, re
from datetime import datetime
from typing import Any, List
import numpy as np
import pandas as pd


PROJECTS_ROOT = "/content/drive/My Drive/2026/tenet/SSOT/RFQI v2/submissions/"
REGISTRY_PATH = os.path.join(PROJECTS_ROOT, "submission_registry.json")


INVERTER_DIR = "/content/drive/My Drive/2026/tenet/SSOT/RFQI v2/component_files/"
INVERTER_FILE = "tenet_hmx_unified_inverter_schema_v1.xlsx"
INVERTER_PATH = os.path.join(INVERTER_DIR, INVERTER_FILE)
INVERTER_SHEET = 0


def n(v: Any, default: float = 0.0) -> float:
    if v is None:
        return default
    try:
        if pd.isna(v):
            return default
    except Exception:
        pass
    try:
        if isinstance(v, str):
            cleaned = (
                v.replace(",", "")
                 .replace("kWh", "")
                 .replace("kVA", "")
                 .replace("kW", "")
                 .replace("VAC", "")
                 .replace("VDC", "")
                 .replace("V", "")
                 .replace("A", "")
                 .replace("%", "")
                 .strip()
            )
            return float(cleaned)
        return float(v)
    except Exception:
        return default


def s(v: Any) -> str:
    if v is None:
        return ""
    try:
        if pd.isna(v):
            return ""
    except Exception:
        pass
    return str(v).strip()


def norm_text(x: Any) -> str:
    x = s(x).lower()
    x = x.replace("&", "and")
    x = x.replace("-", " ")
    x = x.replace("/", " ")
    x = x.replace("_", " ")
    x = " ".join(x.split())
    return x


def yn(v: Any) -> bool:
    return norm_text(v) in {"yes", "y", "true", "1", "supported", "available"}


def make_json_safe(obj):
    if isinstance(obj, dict):
        return {str(k): make_json_safe(v) for k, v in obj.items()}
    if isinstance(obj, (list, tuple, set)):
        return [make_json_safe(v) for v in obj]
    if isinstance(obj, (pd.Timestamp, datetime)):
        return obj.isoformat()
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.floating):
        return float(obj)
    if isinstance(obj, np.bool_):
        return bool(obj)
    try:
        if pd.isna(obj):
            return None
    except Exception:
        pass
    return obj if isinstance(obj, (str, int, float, bool)) or obj is None else str(obj)


def save_json(path: str, data: dict):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(make_json_safe(data), f, indent=4, ensure_ascii=False)


def load_json(path: str):
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing file: {path}")
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def parse_range_min_max(v: Any):
    txt = s(v)

    if not txt:
        return 0.0, 0.0

    # Normalize range separators without turning second value negative
    txt = (
        txt.replace("–", " to ")
           .replace("—", " to ")
           .replace("-", " to ")
           .replace("±", "")
    )

    nums = re.findall(r"\d*\.?\d+", txt)

    if len(nums) >= 2:
        return float(nums[0]), float(nums[1])

    if len(nums) == 1:
        val = float(nums[0])
        return val, val

    return 0.0, 0.0



def phase_match_score(inv_phase: str, supply_phase: str) -> int:
    ip = norm_text(inv_phase)
    sp = norm_text(supply_phase)
    if not ip or not sp:
        return 0
    if ip == sp:
        return 1
    if "single" in sp and "single" in ip:
        return 1
    if "three" in sp and "three" in ip:
        return 1
    return 0


def topology_priority_score(topology: str) -> int:
    t = norm_text(topology)
    if "hybrid" in t:
        return 3
    if "off grid" in t or "offgrid" in t:
        return 2
    if "grid tie" in t or "grid tied" in t:
        return 1
    return 0


def battery_support_score(v_nom: float, battery_type: str) -> int:
    score = 0
    if v_nom > 0:
        score += 1
    if battery_type:
        score += 1
    return score


def battery_bus_suitability_score(v_nom: float, batt_kwh: float) -> int:
    if batt_kwh <= 0:
        return 0
    if batt_kwh <= 5:
        return 2 if v_nom >= 24 else 1 if v_nom > 0 else 0
    if batt_kwh <= 10:
        return 3 if v_nom >= 48 else 1 if v_nom >= 24 else -2
    return 3 if v_nom >= 48 else -1 if v_nom >= 24 else -3


def surge_compliance_flag(surge_kva: float, required_surge_kw: float) -> bool:
    return surge_kva > 0 and surge_kva >= required_surge_kw


def use_case_bonus(use_case: str, priority: str) -> int:
    txt = norm_text(use_case)
    score = 0
    if "residential" in txt:
        score += 1
    if "hybrid" in txt or "storage" in txt:
        score += 1
    if priority and priority in txt:
        score += 1
    return score


registry = load_json(REGISTRY_PATH)
submissions = registry.get("submissions", {})
if not submissions:
    raise ValueError("No submissions found in registry.")


latest_key = max(submissions.keys(), key=lambda k: submissions[k].get("last_seen_at", ""))
project_root = submissions[latest_key]["project_root"]


load_dir = os.path.join(project_root, "01_load")
design_dir = os.path.join(project_root, "02_design")
os.makedirs(design_dir, exist_ok=True)


load_state_path = os.path.join(load_dir, "load_state.json")
load_state = load_json(load_state_path)


inputs_used = load_state.get("inputs_used", {})
load_results = load_state.get("load_results", {})
battery_outputs = load_state.get("battery_outputs", {})


supply_phase = norm_text(inputs_used.get("supply_phase"))
system_priority = norm_text(inputs_used.get("system_priority"))
required_peak_kw = n(load_results.get("p_peak_run_kw"), 0.0)
required_surge_kw = n(load_results.get("p_surge_design_kw"), 0.0)
battery_balanced_kwh = n(battery_outputs.get("battery_balanced_kwh"), 0.0)


if not os.path.exists(INVERTER_PATH):
    raise FileNotFoundError(f"Inverter file not found: {INVERTER_PATH}")


inv_raw_df = pd.read_excel(INVERTER_PATH, sheet_name=INVERTER_SHEET)


print("Inverter schema loaded:", INVERTER_PATH)
print("Rows:", len(inv_raw_df))
print("Columns:")
for c in inv_raw_df.columns:
    print("-", c)
print()


normalized_rows = []


for idx, r in inv_raw_df.iterrows():
    full_schema = {str(c): make_json_safe(r.get(c)) for c in inv_raw_df.columns}


    brand = s(r.get("brand"))
    series = s(r.get("series"))
    model = s(r.get("model"))


    identifier = " | ".join([x for x in [brand, model] if x]).strip()
    if not identifier:
        identifier = f"row_{idx}"


    mppt_low, mppt_high = parse_range_min_max(r.get("mppt_voltage_range_v"))
    batt_low, batt_high = parse_range_min_max(r.get("battery_voltage_range_v"))


    row = {
        "source_row_index": int(idx),
        "identifier": identifier,
        "brand": brand,
        "series": series,
        "model": model,


        "source_file": s(r.get("source_file")),
        "source_page": s(r.get("source_page")),
        "topology": norm_text(r.get("topology")),
        "phase": norm_text(r.get("phase")),


        "rated_power_kw": n(r.get("rated_power_kw")),
        "rated_power_kva": n(r.get("rated_power_kva")),
        "surge_power_kva": n(r.get("surge_power_kva")),


        "ac_output_voltage": s(r.get("ac_output_voltage")),
        "ac_output_frequency_hz": n(r.get("ac_output_frequency_hz")),
        "ac_input_voltage_range": s(r.get("ac_input_voltage_range")),
        "transfer_time_ms": s(r.get("transfer_time_ms")),
        "power_factor": n(r.get("power_factor")),
        "grid_type": s(r.get("grid_type")),


        "battery_voltage_nominal_v": n(r.get("battery_voltage_nominal_v")),
        "battery_voltage_range_v": s(r.get("battery_voltage_range_v")),
        "battery_voltage_min_v": batt_low,
        "battery_voltage_max_v": batt_high,
        "battery_type_supported": norm_text(r.get("battery_type_supported")),
        "max_charge_current_a": n(r.get("max_charge_current_a")),
        "max_discharge_current_a": n(r.get("max_discharge_current_a")),
        "battery_communication": s(r.get("battery_communication")),
        "bms_support": s(r.get("bms_support")),


        "mppt_count": n(r.get("mppt_count")),
        "max_pv_input_power_w": n(r.get("max_pv_input_power_w")),
        "max_pv_input_voltage_v": n(r.get("max_pv_input_voltage_v")),
        "mppt_voltage_range_v": s(r.get("mppt_voltage_range_v")),
        "mppt_voltage_min_v": mppt_low,
        "mppt_voltage_max_v": mppt_high,
        "max_pv_input_current_a": n(r.get("max_pv_input_current_a")),
        "max_pv_short_circuit_current_a": n(r.get("max_pv_short_circuit_current_a")),
        "max_efficiency_percent": n(r.get("max_efficiency_percent")),
        "mppt_efficiency_percent": n(r.get("mppt_efficiency_percent")),


        "parallel_capability": yn(r.get("parallel_capability")),
        "max_parallel_units": n(r.get("max_parallel_units")),
        "ac_coupling": s(r.get("ac_coupling")),
        "generator_support": s(r.get("generator_support")),
        "unbalanced_output_support": s(r.get("unbalanced_output_support")),


        "ip_rating": s(r.get("ip_rating")),
        "cooling_method": s(r.get("cooling_method")),
        "operating_temperature_range_c": s(r.get("operating_temperature_range_c")),
        "dimensions_mm": s(r.get("dimensions_mm")),
        "weight_kg": n(r.get("weight_kg")),
        "display_type": s(r.get("display_type")),
        "communication_ports": s(r.get("communication_ports")),
        "monitoring": s(r.get("monitoring")),
        "recommended_use_case": norm_text(r.get("recommended_use_case")),
        "battery_architecture": norm_text(r.get("battery_architecture")),
        "installation_type": s(r.get("installation_type")),
        "notes": s(r.get("notes")),


        "full_schema_row": full_schema,
    }


    normalized_rows.append(row)


inv_df = pd.DataFrame(normalized_rows)


candidates = []


for _, r in inv_df.iterrows():
    phase_ok = phase_match_score(r["phase"], supply_phase) == 1
    battery_ok = battery_support_score(r["battery_voltage_nominal_v"], r["battery_type_supported"]) > 0
    power_ok = r["rated_power_kw"] >= required_peak_kw
    surge_ok = surge_compliance_flag(r["surge_power_kva"], required_surge_kw)


    if not phase_ok or not battery_ok or not power_ok:
        continue


    oversize_kw = max(0.0, r["rated_power_kw"] - required_peak_kw)


    score = 0
    score += 50 * phase_match_score(r["phase"], supply_phase)
    score += 20 * topology_priority_score(r["topology"])
    score += 20 if surge_ok else -10
    score += 10 * battery_support_score(r["battery_voltage_nominal_v"], r["battery_type_supported"])
    score += 10 * battery_bus_suitability_score(r["battery_voltage_nominal_v"], battery_balanced_kwh)
    score += 5 if r["parallel_capability"] else 0
    score += 2 * use_case_bonus(r["recommended_use_case"], system_priority)
    score -= oversize_kw * 3.0


    candidate = r.to_dict()
    candidate.update({
        "phase_ok": phase_ok,
        "battery_ok": battery_ok,
        "power_ok": power_ok,
        "surge_ok": surge_ok,
        "oversize_kw": oversize_kw,
        "score": score,
    })
    candidates.append(candidate)


candidates_df = pd.DataFrame(candidates)
if len(candidates_df):
    candidates_df = candidates_df.sort_values(
        ["score", "rated_power_kw"],
        ascending=[False, True]
    ).reset_index(drop=True)


rejections = []


for _, r in inv_df.iterrows():
    phase_ok = phase_match_score(r["phase"], supply_phase) == 1
    battery_ok = battery_support_score(r["battery_voltage_nominal_v"], r["battery_type_supported"]) > 0
    power_ok = r["rated_power_kw"] >= required_peak_kw
    surge_ok = surge_compliance_flag(r["surge_power_kva"], required_surge_kw)


    reasons = []
    if not phase_ok:
        reasons.append("phase_mismatch")
    if not battery_ok:
        reasons.append("no_battery_support")
    if not power_ok:
        reasons.append("insufficient_rated_power")
    if power_ok and not surge_ok:
        reasons.append("surge_not_met")


    if reasons:
        rejections.append({
            "identifier": r["identifier"],
            "brand": r["brand"],
            "model": r["model"],
            "phase": r["phase"],
            "topology": r["topology"],
            "rated_power_kw": r["rated_power_kw"],
            "surge_power_kva": r["surge_power_kva"],
            "battery_voltage_nominal_v": r["battery_voltage_nominal_v"],
            "max_pv_input_power_w": r["max_pv_input_power_w"],
            "max_pv_input_voltage_v": r["max_pv_input_voltage_v"],
            "mppt_voltage_range_v": r["mppt_voltage_range_v"],
            "mppt_voltage_min_v": r["mppt_voltage_min_v"],
            "mppt_voltage_max_v": r["mppt_voltage_max_v"],
            "max_pv_input_current_a": r["max_pv_input_current_a"],
            "max_pv_short_circuit_current_a": r["max_pv_short_circuit_current_a"],
            "reasons": reasons,
        })


if len(candidates_df) == 0:
    selected_inverter = {}
    shortlist = []
    selection_status = "no_candidate_found"
else:
    selected_inverter = candidates_df.iloc[0].to_dict()
    shortlist = candidates_df.head(5).to_dict(orient="records")
    selection_status = "selected"


selection_reasons = []


if selected_inverter:
    selection_reasons = [
        "Phase matched required supply.",
        "Battery support available.",
        "Rated power meets or exceeds peak-with-event requirement.",
        "Surge capability meets or exceeds surge design requirement." if selected_inverter.get("surge_ok") else "Surge capability not explicit or below target.",
        "Hybrid topology prioritized for battery-backed architecture." if "hybrid" in norm_text(selected_inverter.get("topology")) else "Non-hybrid topology retained based on score.",
        "PV-side constraints carried forward for downstream PV string design.",
        "Full selected inverter schema row preserved in JSON handoff.",
    ]


selection_state = {
    "meta": {
        "generated_at": datetime.now().isoformat(),
        "submission_key": latest_key,
        "project_root": project_root,
        "source_load_state": load_state_path,
        "source_inverter_schema": INVERTER_PATH,
    },
    "selection_inputs": {
        "supply_phase": supply_phase,
        "system_priority": system_priority,
        "required_peak_kw": required_peak_kw,
        "required_surge_kw": required_surge_kw,
        "battery_balanced_kwh": battery_balanced_kwh,
    },
    "selection_status": selection_status,
    "selected_inverter": selected_inverter,
    "shortlist": shortlist,
    "rejections": rejections,
    "selection_reasons": selection_reasons,
}


selection_state_path = os.path.join(design_dir, "design_state_step_01_inverter.json")
candidate_debug_path = os.path.join(design_dir, "inverter_candidates_debug.json")
selection_report_md_path = os.path.join(design_dir, "design_step_01_inverter_report.md")


save_json(selection_state_path, selection_state)


save_json(candidate_debug_path, {
    "schema_columns": list(inv_raw_df.columns),
    "normalized_preview": inv_df.head(20).to_dict(orient="records"),
    "candidate_count": int(len(candidates_df)),
    "top_candidates_preview": shortlist,
    "rejections_preview": rejections[:20],
})


if selected_inverter:
    selected_md = f"""
## Selected Inverter
- Identifier: **{selected_inverter.get('identifier','')}**
- Brand: **{selected_inverter.get('brand','')}**
- Series: **{selected_inverter.get('series','')}**
- Model: **{selected_inverter.get('model','')}**
- Topology: **{selected_inverter.get('topology','')}**
- Phase: **{selected_inverter.get('phase','')}**
- Rated Power: **{n(selected_inverter.get('rated_power_kw'),0):.2f} kW**
- Surge Power: **{n(selected_inverter.get('surge_power_kva'),0):.2f} kVA**
- Battery Voltage Nominal: **{n(selected_inverter.get('battery_voltage_nominal_v'),0):.0f} V**
- Battery Voltage Range: **{selected_inverter.get('battery_voltage_range_v','')}**
- MPPT Count: **{n(selected_inverter.get('mppt_count'),0):.0f}**
- Max PV Input Power: **{n(selected_inverter.get('max_pv_input_power_w'),0):.0f} W**
- Max PV Input Voltage: **{n(selected_inverter.get('max_pv_input_voltage_v'),0):.0f} V**
- MPPT Voltage Range: **{selected_inverter.get('mppt_voltage_range_v','')}**
- MPPT Min Voltage: **{n(selected_inverter.get('mppt_voltage_min_v'),0):.0f} V**
- MPPT Max Voltage: **{n(selected_inverter.get('mppt_voltage_max_v'),0):.0f} V**
- Max PV Input Current: **{n(selected_inverter.get('max_pv_input_current_a'),0):.2f} A**
- Max PV Short-Circuit Current: **{n(selected_inverter.get('max_pv_short_circuit_current_a'),0):.2f} A**
- Max Efficiency: **{n(selected_inverter.get('max_efficiency_percent'),0):.2f}%**
- MPPT Efficiency: **{n(selected_inverter.get('mppt_efficiency_percent'),0):.2f}%**
- Parallel Capability: **{selected_inverter.get('parallel_capability')}**
- Max Parallel Units: **{n(selected_inverter.get('max_parallel_units'),0):.0f}**
- Recommended Use Case: **{selected_inverter.get('recommended_use_case','')}**
- Battery Architecture: **{selected_inverter.get('battery_architecture','')}**
- Full Schema Row Preserved: **Yes**
"""
else:
    selected_md = "## Selected Inverter\n- No inverter selected.\n"


shortlist_md = "\n".join([
    f"- **{i+1}. {r.get('identifier','')}** | "
    f"Rated: {n(r.get('rated_power_kw'),0):.2f} kW | "
    f"PV Max: {n(r.get('max_pv_input_power_w'),0):.0f} W | "
    f"Vdc Max: {n(r.get('max_pv_input_voltage_v'),0):.0f} V | "
    f"MPPT: {n(r.get('mppt_voltage_min_v'),0):.0f}-{n(r.get('mppt_voltage_max_v'),0):.0f} V | "
    f"Ipv Max: {n(r.get('max_pv_input_current_a'),0):.2f} A | "
    f"Score: {n(r.get('score'),0):.2f}"
    for i, r in enumerate(shortlist)
]) if shortlist else "- No compliant candidates found."


selection_reasons_md = "\n".join([f"- {x}" for x in selection_reasons]) if selection_reasons else "- No selection reasons available."


report_md = f"""# TENET Design Step 01 — Inverter Selection Report


## Selection Inputs
- Supply Phase: **{supply_phase}**
- System Priority: **{system_priority}**
- Required Peak with Event: **{required_peak_kw:.2f} kW**
- Required Surge: **{required_surge_kw:.2f} kW~kVA**
- Balanced Battery Target: **{battery_balanced_kwh:.2f} kWh**


{selected_md}


## Selection Rationale
{selection_reasons_md}


## Shortlist
{shortlist_md}


## Canonical Handoff
- JSON State: `{selection_state_path}`
- Candidate Debug: `{candidate_debug_path}`
"""


with open(selection_report_md_path, "w", encoding="utf-8") as f:
    f.write(report_md)


print("----- BLOCK 3 COMPLETE -----")
print("Selection status:", selection_status)
print("JSON handoff:", selection_state_path)
print("Markdown report:", selection_report_md_path)
print("Candidate debug:", candidate_debug_path)
print()


if selected_inverter:
    print("Selected inverter:", selected_inverter.get("identifier"))
    print("Rated power (kW):", round(n(selected_inverter.get("rated_power_kw"), 0), 2))
    print("Surge power (kVA):", round(n(selected_inverter.get("surge_power_kva"), 0), 2))
    print("Battery bus (V):", round(n(selected_inverter.get("battery_voltage_nominal_v"), 0), 0))
    print("Max PV input power (W):", round(n(selected_inverter.get("max_pv_input_power_w"), 0), 0))
    print("Max PV input voltage (V):", round(n(selected_inverter.get("max_pv_input_voltage_v"), 0), 0))
    print("MPPT voltage min/max (V):", round(n(selected_inverter.get("mppt_voltage_min_v"), 0), 0), "/", round(n(selected_inverter.get("mppt_voltage_max_v"), 0), 0))
    print("Max PV input current (A):", round(n(selected_inverter.get("max_pv_input_current_a"), 0), 2))
    print("Max PV short-circuit current (A):", round(n(selected_inverter.get("max_pv_short_circuit_current_a"), 0), 2))
    print("Full schema fields carried:", len(selected_inverter.get("full_schema_row", {})))
else:
    print("No inverter selected.")

```

```python
# =========================
# BLOCK 4: BATTERY SELECTION
# FULL REVISED SCRIPT
# - Uses full inverter handoff
# - Uses nominal kWh for product selection
# - Keeps usable kWh for discharge/autonomy reference
# - Prioritizes 1 battery module where within tolerance
# - Preserves full selected battery schema row
# =========================


import os, json, math
from datetime import datetime
from typing import Any, List
import numpy as np
import pandas as pd


PROJECTS_ROOT = "/content/drive/My Drive/2026/tenet/SSOT/RFQI v2/submissions/"
REGISTRY_PATH = os.path.join(PROJECTS_ROOT, "submission_registry.json")


BATTERY_DIR = "/content/drive/My Drive/2026/tenet/SSOT/RFQI v2/component_files/"
BATTERY_FILE = "tenet_hmx_unified_battery_schema_v1.xlsx"
BATTERY_PATH = os.path.join(BATTERY_DIR, BATTERY_FILE)
BATTERY_SHEET = 0


BATTERY_USABLE_DOD = 0.90
SINGLE_UNIT_SHORTFALL_TOLERANCE_PCT = 10.0
MAX_SEARCH_UNITS = 16


def n(v: Any, default: float = 0.0) -> float:
    if v is None:
        return default
    try:
        if pd.isna(v):
            return default
    except Exception:
        pass
    try:
        if isinstance(v, str):
            cleaned = (
                v.replace(",", "")
                 .replace("kWh", "")
                 .replace("kW", "")
                 .replace("V", "")
                 .replace("Ah", "")
                 .replace("%", "")
                 .strip()
            )
            return float(cleaned)
        return float(v)
    except Exception:
        return default


def s(v: Any) -> str:
    if v is None:
        return ""
    try:
        if pd.isna(v):
            return ""
    except Exception:
        pass
    return str(v).strip()


def norm_text(x: Any) -> str:
    x = s(x).lower()
    x = x.replace("&", "and")
    x = x.replace("-", " ")
    x = x.replace("/", " ")
    x = x.replace("_", " ")
    x = " ".join(x.split())
    return x


def yn(v: Any) -> bool:
    return norm_text(v) in {"yes", "y", "true", "1", "supported", "available"}


def make_json_safe(obj):
    if isinstance(obj, dict):
        return {str(k): make_json_safe(v) for k, v in obj.items()}
    if isinstance(obj, (list, tuple, set)):
        return [make_json_safe(v) for v in obj]
    if isinstance(obj, (pd.Timestamp, datetime)):
        return obj.isoformat()
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.floating):
        return float(obj)
    if isinstance(obj, np.bool_):
        return bool(obj)
    if isinstance(obj, np.ndarray):
        return [make_json_safe(v) for v in obj.tolist()]
    try:
        if pd.isna(obj):
            return None
    except Exception:
        pass
    return obj if isinstance(obj, (str, int, float, bool)) or obj is None else str(obj)


def save_json(path: str, data: dict):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(make_json_safe(data), f, indent=4, ensure_ascii=False)


def load_json(path: str):
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing file: {path}")
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def pick_first_existing(row: pd.Series, candidates: List[str], default=""):
    for c in candidates:
        if c in row.index:
            val = row.get(c)
            try:
                if pd.isna(val):
                    continue
            except Exception:
                pass
            return val
    return default


def voltage_compatible(inv_nom_v, inv_min_v, inv_max_v, bat_v, inv_arch, bat_arch, bat_method, notes):
    inv_arch = norm_text(inv_arch)
    bat_arch = norm_text(bat_arch)
    bat_method = norm_text(bat_method)
    notes = norm_text(notes)


    if bat_v <= 0:
        return False, "invalid_battery_voltage"


    if inv_nom_v <= 60:
        if "25.6v version" in notes or abs(bat_v - 25.6) < 0.2:
            return False, "25_6v_variant_not_suitable_for_48v_inverter"


        if not (40 <= bat_v <= 60):
            return False, "not_48v_low_voltage_class"


        if abs(bat_v - inv_nom_v) < 0.5:
            return True, "exact_nominal_voltage_match"


        if abs(bat_v - 51.2) < 0.5 and abs(inv_nom_v - 48) < 1.0:
            return True, "51_2v_battery_on_48v_inverter_class"


        pct_diff = abs(bat_v - inv_nom_v) / max(inv_nom_v, 1e-9)
        if pct_diff <= 0.20:
            return True, "low_voltage_class_compatible"


        return False, "low_voltage_nominal_mismatch"


    return False, "high_voltage_battery_logic_not_enabled"


# -------------------------------------------------
# FIND CURRENT SUBMISSION
# -------------------------------------------------
registry = load_json(REGISTRY_PATH)
submissions = registry.get("submissions", {})
if not submissions:
    raise ValueError("No submissions found in registry.")


latest_key = max(submissions.keys(), key=lambda k: submissions[k].get("last_seen_at", ""))
project_root = submissions[latest_key]["project_root"]


design_dir = os.path.join(project_root, "02_design")
load_dir = os.path.join(project_root, "01_load")
os.makedirs(design_dir, exist_ok=True)


inverter_state_path = os.path.join(design_dir, "design_state_step_01_inverter.json")
load_state_path = os.path.join(load_dir, "load_state.json")


inverter_state = load_json(inverter_state_path)
load_state = load_json(load_state_path)


selected_inverter = inverter_state.get("selected_inverter", {})
battery_outputs = load_state.get("battery_outputs", {})


inv_identifier = s(selected_inverter.get("identifier"))
inv_battery_voltage_v = n(selected_inverter.get("battery_voltage_nominal_v"), 0.0)
inv_battery_voltage_min_v = n(selected_inverter.get("battery_voltage_min_v"), 0.0)
inv_battery_voltage_max_v = n(selected_inverter.get("battery_voltage_max_v"), 0.0)
inv_battery_architecture = norm_text(selected_inverter.get("battery_architecture"))
inv_battery_type_supported = norm_text(selected_inverter.get("battery_type_supported"))
inv_max_charge_current_a = n(selected_inverter.get("max_charge_current_a"), 0.0)
inv_max_discharge_current_a = n(selected_inverter.get("max_discharge_current_a"), 0.0)
inv_battery_communication = s(selected_inverter.get("battery_communication"))
inv_bms_support = s(selected_inverter.get("bms_support"))


battery_minimum_kwh = n(battery_outputs.get("battery_minimum_kwh"), 0.0)
battery_balanced_kwh = n(battery_outputs.get("battery_balanced_kwh"), 0.0)
battery_future_kwh = n(battery_outputs.get("battery_future_kwh"), 0.0)


print("Current submission key:", latest_key)
print("Project root:", project_root)
print("Selected inverter:", inv_identifier)
print("Inverter battery bus:", inv_battery_voltage_v)
print("Inverter battery architecture:", inv_battery_architecture)
print("Battery minimum / balanced / future kWh:", battery_minimum_kwh, battery_balanced_kwh, battery_future_kwh)
print()


# -------------------------------------------------
# LOAD BATTERY SCHEMA
# -------------------------------------------------
if not os.path.exists(BATTERY_PATH):
    raise FileNotFoundError(f"Battery schema file not found: {BATTERY_PATH}")


bat_raw_df = pd.read_excel(BATTERY_PATH, sheet_name=BATTERY_SHEET)


print("Battery schema loaded:", BATTERY_PATH)
print("Rows:", len(bat_raw_df))
print("Columns:", list(bat_raw_df.columns))
print()


# -------------------------------------------------
# NORMALIZE BATTERY SCHEMA
# -------------------------------------------------
normalized_rows = []


for idx, r in bat_raw_df.iterrows():
    full_schema = {str(c): make_json_safe(r.get(c)) for c in bat_raw_df.columns}


    brand = s(pick_first_existing(r, ["brand", "Brand"]))
    model = s(pick_first_existing(r, ["model", "Model", "battery_model"]))
    series = s(pick_first_existing(r, ["series", "Series"]))
    notes = s(pick_first_existing(r, ["notes", "Notes"]))


    identifier = " | ".join([x for x in [brand, model, series] if x]).strip()
    if not identifier:
        identifier = f"row_{idx}"


    nominal_voltage_v = n(pick_first_existing(r, ["nominal_voltage_v", "Nominal Voltage (V)"]))
    nominal_capacity_ah = n(pick_first_existing(r, ["nominal_capacity_ah", "Nominal Capacity (Ah)"]))
    nominal_energy_kwh = n(pick_first_existing(r, ["nominal_energy_kwh", "Nominal Energy (kWh)"]))


    if nominal_energy_kwh <= 0 and nominal_voltage_v > 0 and nominal_capacity_ah > 0:
        nominal_energy_kwh = (nominal_voltage_v * nominal_capacity_ah) / 1000.0


    chemistry = norm_text(pick_first_existing(r, ["chemistry", "Chemistry"]))
    battery_matching_method = norm_text(pick_first_existing(r, ["battery_matching_method", "Battery Matching Method"]))
    battery_architecture = norm_text(pick_first_existing(r, ["battery_architecture", "Battery Architecture"]))
    installation_type = norm_text(pick_first_existing(r, ["installation_type", "Installation Type"]))


    supports_parallel = True
    max_parallel_units = MAX_SEARCH_UNITS


    normalized_rows.append({
        "source_row_index": int(idx),
        "identifier": identifier,
        "brand": brand,
        "model": model,
        "series": series,
        "nominal_voltage_v": nominal_voltage_v,
        "nominal_capacity_ah": nominal_capacity_ah,
        "nominal_energy_kwh": nominal_energy_kwh,
        "usable_energy_kwh": nominal_energy_kwh * BATTERY_USABLE_DOD,
        "chemistry": chemistry,
        "battery_matching_method": battery_matching_method,
        "battery_architecture": battery_architecture,
        "installation_type": installation_type,
        "supports_parallel": supports_parallel,
        "max_parallel_units": max_parallel_units,
        "notes": notes,
        "full_schema_row": full_schema,
    })


bat_df = pd.DataFrame(normalized_rows)


print("Normalized battery rows:")
print(bat_df[[
    "identifier",
    "nominal_voltage_v",
    "nominal_capacity_ah",
    "nominal_energy_kwh",
    "usable_energy_kwh",
    "battery_architecture"
]].to_string(index=False))
print()


# -------------------------------------------------
# COMPATIBILITY + SELECTION
# -------------------------------------------------
compatible_rows = []
rejections = []


for _, r in bat_df.iterrows():
    bat_v = n(r["nominal_voltage_v"])
    bat_nom_kwh = n(r["nominal_energy_kwh"])
    bat_usable_kwh = n(r["usable_energy_kwh"])
    max_parallel_units = int(n(r["max_parallel_units"], MAX_SEARCH_UNITS))
    supports_parallel = bool(r["supports_parallel"])


    if bat_v <= 0 or bat_nom_kwh <= 0:
        rejections.append({**r.to_dict(), "reasons": ["invalid_voltage_or_nominal_energy"]})
        continue


    ok, voltage_quality = voltage_compatible(
        inv_battery_voltage_v,
        inv_battery_voltage_min_v,
        inv_battery_voltage_max_v,
        bat_v,
        inv_battery_architecture,
        r["battery_architecture"],
        r["battery_matching_method"],
        r["notes"],
    )


    if not ok:
        rejections.append({**r.to_dict(), "reasons": [voltage_quality]})
        continue


    if not supports_parallel:
        rejections.append({**r.to_dict(), "reasons": ["parallel_not_supported"]})
        continue


    # Product selection uses nominal kWh
    selection_energy_kwh = bat_nom_kwh


    # Single-unit balanced test
    single_delivered_kwh = selection_energy_kwh
    single_shortfall_kwh = max(0.0, battery_balanced_kwh - single_delivered_kwh)
    single_excess_kwh = max(0.0, single_delivered_kwh - battery_balanced_kwh)
    single_variance_kwh = abs(single_delivered_kwh - battery_balanced_kwh)


    single_shortfall_pct = (
        single_shortfall_kwh / battery_balanced_kwh * 100.0
        if battery_balanced_kwh > 0 else 0.0
    )


    single_unit_within_tolerance = (
        single_shortfall_pct <= SINGLE_UNIT_SHORTFALL_TOLERANCE_PCT
    )


    # Multi-unit search only for ranking/reference
    unit_options = []


    for units in range(1, max_parallel_units + 1):
        delivered_kwh = units * selection_energy_kwh
        usable_delivered_kwh = units * bat_usable_kwh
        shortfall_kwh = max(0.0, battery_balanced_kwh - delivered_kwh)
        excess_kwh = max(0.0, delivered_kwh - battery_balanced_kwh)
        variance_kwh = abs(delivered_kwh - battery_balanced_kwh)


        shortfall_pct = shortfall_kwh / battery_balanced_kwh * 100.0 if battery_balanced_kwh > 0 else 0.0
        excess_pct = excess_kwh / battery_balanced_kwh * 100.0 if battery_balanced_kwh > 0 else 0.0
        variance_pct = variance_kwh / battery_balanced_kwh * 100.0 if battery_balanced_kwh > 0 else 0.0


        meets_bal_nominal = delivered_kwh >= battery_balanced_kwh
        within_tolerance = shortfall_pct <= SINGLE_UNIT_SHORTFALL_TOLERANCE_PCT


        score = 0
        score += 1000 if meets_bal_nominal else 0
        score += 700 if units == 1 and within_tolerance else 0
        score -= variance_pct * 20
        score -= excess_pct * 5
        score -= units * 15


        unit_options.append({
            "units": units,
            "delivered_nominal_kwh": delivered_kwh,
            "delivered_usable_kwh": usable_delivered_kwh,
            "shortfall_kwh": shortfall_kwh,
            "excess_kwh": excess_kwh,
            "variance_kwh": variance_kwh,
            "shortfall_pct": shortfall_pct,
            "excess_pct": excess_pct,
            "variance_pct": variance_pct,
            "meets_bal_nominal": meets_bal_nominal,
            "within_tolerance": within_tolerance,
            "score": score,
        })


    if single_unit_within_tolerance:
        selected_option = unit_options[0]
        selection_mode = "single_unit_within_balanced_tolerance"
    else:
        qualifying = [x for x in unit_options if x["meets_bal_nominal"]]
        if qualifying:
            selected_option = sorted(
                qualifying,
                key=lambda x: (x["variance_pct"], x["excess_pct"], x["units"])
            )[0]
            selection_mode = "nearest_nominal_meeting_balanced"
        else:
            selected_option = sorted(
                unit_options,
                key=lambda x: (x["variance_pct"], x["shortfall_pct"], x["units"])
            )[0]
            selection_mode = "nearest_nominal_even_if_short"


    candidate_score = selected_option["score"]


    if selected_option["units"] == 1:
        candidate_score += 500


    if voltage_quality == "exact_nominal_voltage_match":
        candidate_score += 30
    elif voltage_quality == "51_2v_battery_on_48v_inverter_class":
        candidate_score += 25
    else:
        candidate_score += 10


    compatible_rows.append({
        **r.to_dict(),
        "selection_energy_basis": "nominal_energy_kwh",
        "selection_energy_kwh": selection_energy_kwh,
        "usable_energy_reference_kwh": bat_usable_kwh,
        "voltage_match_quality": voltage_quality,
        "selection_mode": selection_mode,


        "single_unit_delivered_nominal_kwh": single_delivered_kwh,
        "single_unit_shortfall_kwh": single_shortfall_kwh,
        "single_unit_excess_kwh": single_excess_kwh,
        "single_unit_variance_kwh": single_variance_kwh,
        "single_unit_shortfall_pct": single_shortfall_pct,
        "single_unit_within_tolerance": int(single_unit_within_tolerance),


        "selected_units_bal": int(selected_option["units"]),
        "selected_delivered_balanced_kwh": selected_option["delivered_nominal_kwh"],
        "selected_delivered_usable_kwh": selected_option["delivered_usable_kwh"],
        "selected_shortfall_kwh": selected_option["shortfall_kwh"],
        "selected_excess_kwh": selected_option["excess_kwh"],
        "selected_variance_kwh": selected_option["variance_kwh"],
        "selected_shortfall_pct": selected_option["shortfall_pct"],
        "selected_excess_pct": selected_option["excess_pct"],
        "selected_variance_pct": selected_option["variance_pct"],
        "selected_meets_bal_nominal": int(selected_option["meets_bal_nominal"]),
        "selected_within_tolerance": int(selected_option["within_tolerance"]),


        "unit_options": unit_options,
        "score": candidate_score,
    })


if compatible_rows:
    compatible_df = pd.DataFrame(compatible_rows).sort_values(
        [
            "single_unit_within_tolerance",
            "selected_units_bal",
            "selected_variance_pct",
            "selected_excess_pct",
            "score",
        ],
        ascending=[False, True, True, True, False]
    ).reset_index(drop=True)
else:
    compatible_df = pd.DataFrame()


print("Compatible battery candidates:", len(compatible_df))
if len(compatible_df):
    print(compatible_df[[
        "identifier",
        "nominal_voltage_v",
        "nominal_energy_kwh",
        "usable_energy_kwh",
        "selection_mode",
        "selected_units_bal",
        "selected_delivered_balanced_kwh",
        "selected_variance_pct",
        "single_unit_within_tolerance",
        "score"
    ]].to_string(index=False))
else:
    print("No compatible battery candidates found.")


print()


# -------------------------------------------------
# SELECT + SAVE
# -------------------------------------------------
if len(compatible_df):
    selected_battery = compatible_df.iloc[0].to_dict()
    shortlist = compatible_df.head(10).to_dict(orient="records")
    selection_status = "selected"
else:
    selected_battery = {}
    shortlist = []
    selection_status = "no_candidate_found"


battery_state_path = os.path.join(design_dir, "design_state_step_02_battery.json")
battery_report_md_path = os.path.join(design_dir, "design_step_02_battery_report.md")
battery_debug_path = os.path.join(design_dir, "battery_candidates_debug.json")


battery_state = {
    "meta": {
        "generated_at": datetime.now().isoformat(),
        "submission_key": latest_key,
        "project_root": project_root,
        "source_inverter_state": inverter_state_path,
        "source_load_state": load_state_path,
        "source_battery_schema": BATTERY_PATH,
    },
    "selection_inputs": {
        "selected_inverter_identifier": inv_identifier,
        "inverter_battery_voltage_nominal_v": inv_battery_voltage_v,
        "inverter_battery_voltage_min_v": inv_battery_voltage_min_v,
        "inverter_battery_voltage_max_v": inv_battery_voltage_max_v,
        "inverter_battery_architecture": inv_battery_architecture,
        "inverter_battery_type_supported": inv_battery_type_supported,
        "inverter_max_charge_current_a": inv_max_charge_current_a,
        "inverter_max_discharge_current_a": inv_max_discharge_current_a,
        "inverter_battery_communication": inv_battery_communication,
        "inverter_bms_support": inv_bms_support,
        "battery_minimum_kwh": battery_minimum_kwh,
        "battery_balanced_kwh": battery_balanced_kwh,
        "battery_future_kwh": battery_future_kwh,
        "battery_usable_dod_reference": BATTERY_USABLE_DOD,
        "single_unit_shortfall_tolerance_pct": SINGLE_UNIT_SHORTFALL_TOLERANCE_PCT,
    },
    "selection_status": selection_status,
    "selected_battery": selected_battery,
    "shortlist": shortlist,
}


save_json(battery_state_path, battery_state)


save_json(battery_debug_path, {
    "normalized_rows": bat_df.to_dict(orient="records"),
    "compatible_count": int(len(compatible_df)),
    "compatible_preview": compatible_df.head(20).to_dict(orient="records") if len(compatible_df) else [],
    "rejections_preview": rejections[:100],
})


if selected_battery:
    report_text = f"""# TENET Design Step 02 — Battery Selection Report


## Selected Battery
- Identifier: **{selected_battery.get('identifier','')}**
- Nominal Voltage: **{n(selected_battery.get('nominal_voltage_v'),0):.2f} V**
- Nominal Energy per Unit: **{n(selected_battery.get('nominal_energy_kwh'),0):.4f} kWh**
- Usable Energy per Unit Reference at {BATTERY_USABLE_DOD:.0%} DoD: **{n(selected_battery.get('usable_energy_kwh'),0):.4f} kWh**
- Selection Mode: **{selected_battery.get('selection_mode','')}**
- Selected Units for Balanced Target: **{int(n(selected_battery.get('selected_units_bal'),0))}**
- Selected Delivered Balanced Nominal Energy: **{n(selected_battery.get('selected_delivered_balanced_kwh'),0):.4f} kWh**
- Selected Delivered Usable Energy Reference: **{n(selected_battery.get('selected_delivered_usable_kwh'),0):.4f} kWh**
- Selected Variance from Balanced Target: **{n(selected_battery.get('selected_variance_pct'),0):.2f}%**
- Selected Shortfall from Balanced Target: **{n(selected_battery.get('selected_shortfall_pct'),0):.2f}%**
- Single Unit Within Tolerance: **{selected_battery.get('single_unit_within_tolerance')}**
- Voltage Match Quality: **{selected_battery.get('voltage_match_quality','')}**


## Targets
- Battery Minimum: **{battery_minimum_kwh:.4f} kWh**
- Battery Balanced: **{battery_balanced_kwh:.4f} kWh**
- Battery Future: **{battery_future_kwh:.4f} kWh**


## Canonical Handoff
- JSON State: `{battery_state_path}`
- Debug File: `{battery_debug_path}`
"""
else:
    report_text = f"""# TENET Design Step 02 — Battery Selection Report


No compatible battery candidate found.


## Canonical Handoff
- JSON State: `{battery_state_path}`
- Debug File: `{battery_debug_path}`
"""


with open(battery_report_md_path, "w", encoding="utf-8") as f:
    f.write(report_text)


print()
print("----- BLOCK 4 COMPLETE -----")
print("Selection status:", selection_status)
print("JSON handoff:", battery_state_path)
print("Markdown report:", battery_report_md_path)
print("Debug:", battery_debug_path)


if selected_battery:
    print()
    print("Selected battery:", selected_battery.get("identifier"))
    print("Selection mode:", selected_battery.get("selection_mode"))
    print("Nominal energy per unit (kWh):", round(n(selected_battery.get("nominal_energy_kwh"), 0), 4))
    print("Selected units:", int(n(selected_battery.get("selected_units_bal"), 0)))
    print("Delivered nominal balanced kWh:", round(n(selected_battery.get("selected_delivered_balanced_kwh"), 0), 4))
    print("Variance %:", round(n(selected_battery.get("selected_variance_pct"), 0), 2))

```

```python
# =========================
# BLOCK 4: BATTERY SELECTION
# FULL REVISED SCRIPT
# Clean selected fields + renamed ceil_reference fields
# =========================


import os, json, math
from datetime import datetime
from typing import Any, List
import numpy as np
import pandas as pd


PROJECTS_ROOT = "/content/drive/My Drive/2026/tenet/SSOT/RFQI v2/submissions/"
REGISTRY_PATH = os.path.join(PROJECTS_ROOT, "submission_registry.json")


BATTERY_DIR = "/content/drive/My Drive/2026/tenet/SSOT/RFQI v2/component_files/"
BATTERY_FILE = "tenet_hmx_unified_battery_schema_v1.xlsx"
BATTERY_PATH = os.path.join(BATTERY_DIR, BATTERY_FILE)
BATTERY_SHEET = 0


BATTERY_USABLE_DOD = 0.90
SINGLE_UNIT_SHORTFALL_TOLERANCE_PCT = 10.0
MAX_SEARCH_UNITS = 16




# -------------------------------------------------
# HELPERS
# -------------------------------------------------
def n(v: Any, default: float = 0.0) -> float:
    if v is None:
        return default
    try:
        if pd.isna(v):
            return default
    except Exception:
        pass
    try:
        if isinstance(v, str):
            cleaned = (
                v.replace(",", "")
                 .replace("kWh", "")
                 .replace("kW", "")
                 .replace("V", "")
                 .replace("Ah", "")
                 .replace("%", "")
                 .strip()
            )
            return float(cleaned)
        return float(v)
    except Exception:
        return default




def s(v: Any) -> str:
    if v is None:
        return ""
    try:
        if pd.isna(v):
            return ""
    except Exception:
        pass
    return str(v).strip()




def norm_text(x: Any) -> str:
    x = s(x).lower()
    x = x.replace("&", "and")
    x = x.replace("-", " ")
    x = x.replace("/", " ")
    x = x.replace("_", " ")
    x = " ".join(x.split())
    return x




def make_json_safe(obj):
    if isinstance(obj, dict):
        return {str(k): make_json_safe(v) for k, v in obj.items()}
    if isinstance(obj, (list, tuple, set)):
        return [make_json_safe(v) for v in obj]
    if isinstance(obj, (pd.Timestamp, datetime)):
        return obj.isoformat()
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.floating):
        return float(obj)
    if isinstance(obj, np.bool_):
        return bool(obj)
    if isinstance(obj, np.ndarray):
        return [make_json_safe(v) for v in obj.tolist()]
    try:
        if pd.isna(obj):
            return None
    except Exception:
        pass
    if isinstance(obj, (str, int, float, bool)) or obj is None:
        return obj
    return str(obj)




def save_json(path: str, data: dict):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(make_json_safe(data), f, indent=4, ensure_ascii=False)




def load_json(path: str):
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing file: {path}")
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)




def pick_first_existing(row: pd.Series, candidates: List[str], default=""):
    for c in candidates:
        if c in row.index:
            val = row.get(c)
            try:
                if pd.isna(val):
                    continue
            except Exception:
                pass
            return val
    return default




def voltage_compatible(inv_nom_v, bat_v, notes):
    notes = norm_text(notes)


    if bat_v <= 0:
        return False, "invalid_battery_voltage"


    if inv_nom_v <= 60:
        if "25.6v version" in notes or abs(bat_v - 25.6) < 0.2:
            return False, "25_6v_variant_not_suitable_for_48v_inverter"


        if not (40 <= bat_v <= 60):
            return False, "not_48v_low_voltage_class"


        if abs(bat_v - inv_nom_v) < 0.5:
            return True, "exact_nominal_voltage_match"


        if abs(bat_v - 51.2) < 0.5 and abs(inv_nom_v - 48) < 1.0:
            return True, "51_2v_battery_on_48v_inverter_class"


        pct_diff = abs(bat_v - inv_nom_v) / max(inv_nom_v, 1e-9)
        if pct_diff <= 0.20:
            return True, "low_voltage_class_compatible"


        return False, "low_voltage_nominal_mismatch"


    return False, "high_voltage_battery_logic_not_enabled"




# -------------------------------------------------
# FIND CURRENT SUBMISSION
# -------------------------------------------------
registry = load_json(REGISTRY_PATH)
submissions = registry.get("submissions", {})
if not submissions:
    raise ValueError("No submissions found in registry.")


latest_key = max(submissions.keys(), key=lambda k: submissions[k].get("last_seen_at", ""))
project_root = submissions[latest_key]["project_root"]


design_dir = os.path.join(project_root, "02_design")
load_dir = os.path.join(project_root, "01_load")
os.makedirs(design_dir, exist_ok=True)


inverter_state_path = os.path.join(design_dir, "design_state_step_01_inverter.json")
load_state_path = os.path.join(load_dir, "load_state.json")


inverter_state = load_json(inverter_state_path)
load_state = load_json(load_state_path)


selected_inverter = inverter_state.get("selected_inverter", {})
battery_outputs = load_state.get("battery_outputs", {})


inv_identifier = s(selected_inverter.get("identifier"))
inv_battery_voltage_v = n(selected_inverter.get("battery_voltage_nominal_v"), 0.0)
inv_battery_architecture = norm_text(selected_inverter.get("battery_architecture"))


battery_minimum_kwh = n(battery_outputs.get("battery_minimum_kwh"), 0.0)
battery_balanced_kwh = n(battery_outputs.get("battery_balanced_kwh"), 0.0)
battery_future_kwh = n(battery_outputs.get("battery_future_kwh"), 0.0)


print("Current submission key:", latest_key)
print("Project root:", project_root)
print("Selected inverter:", inv_identifier)
print("Inverter battery bus:", inv_battery_voltage_v)
print("Battery minimum / balanced / future kWh:", battery_minimum_kwh, battery_balanced_kwh, battery_future_kwh)
print()




# -------------------------------------------------
# LOAD BATTERY SCHEMA
# -------------------------------------------------
if not os.path.exists(BATTERY_PATH):
    raise FileNotFoundError(f"Battery schema file not found: {BATTERY_PATH}")


bat_raw_df = pd.read_excel(BATTERY_PATH, sheet_name=BATTERY_SHEET)


print("Battery schema loaded:", BATTERY_PATH)
print("Rows:", len(bat_raw_df))
print("Columns:", list(bat_raw_df.columns))
print()




# -------------------------------------------------
# NORMALIZE BATTERY SCHEMA
# -------------------------------------------------
normalized_rows = []


for idx, r in bat_raw_df.iterrows():
    full_schema = {str(c): make_json_safe(r.get(c)) for c in bat_raw_df.columns}


    brand = s(pick_first_existing(r, ["brand", "Brand"]))
    model = s(pick_first_existing(r, ["model", "Model", "battery_model"]))
    series = s(pick_first_existing(r, ["series", "Series"]))
    notes = s(pick_first_existing(r, ["notes", "Notes"]))


    identifier = " | ".join([x for x in [brand, model, series] if x]).strip()
    if not identifier:
        identifier = f"row_{idx}"


    nominal_voltage_v = n(pick_first_existing(r, ["nominal_voltage_v", "Nominal Voltage (V)"]))
    nominal_capacity_ah = n(pick_first_existing(r, ["nominal_capacity_ah", "Nominal Capacity (Ah)"]))
    nominal_energy_kwh = n(pick_first_existing(r, ["nominal_energy_kwh", "Nominal Energy (kWh)"]))


    if nominal_energy_kwh <= 0 and nominal_voltage_v > 0 and nominal_capacity_ah > 0:
        nominal_energy_kwh = (nominal_voltage_v * nominal_capacity_ah) / 1000.0


    chemistry = norm_text(pick_first_existing(r, ["chemistry", "Chemistry"]))
    battery_matching_method = norm_text(pick_first_existing(r, ["battery_matching_method", "Battery Matching Method"]))
    battery_architecture = norm_text(pick_first_existing(r, ["battery_architecture", "Battery Architecture"]))
    installation_type = norm_text(pick_first_existing(r, ["installation_type", "Installation Type"]))


    normalized_rows.append({
        "source_row_index": int(idx),
        "identifier": identifier,
        "brand": brand,
        "model": model,
        "series": series,
        "nominal_voltage_v": nominal_voltage_v,
        "nominal_capacity_ah": nominal_capacity_ah,
        "nominal_energy_kwh": nominal_energy_kwh,
        "usable_energy_kwh": nominal_energy_kwh * BATTERY_USABLE_DOD,
        "chemistry": chemistry,
        "battery_matching_method": battery_matching_method,
        "battery_architecture": battery_architecture,
        "installation_type": installation_type,
        "supports_parallel": True,
        "max_parallel_units": MAX_SEARCH_UNITS,
        "notes": notes,
        "full_schema_row": full_schema,
    })


bat_df = pd.DataFrame(normalized_rows)


print("Normalized battery rows:")
print(bat_df[[
    "identifier",
    "nominal_voltage_v",
    "nominal_capacity_ah",
    "nominal_energy_kwh",
    "usable_energy_kwh",
    "battery_architecture"
]].to_string(index=False))
print()




# -------------------------------------------------
# COMPATIBILITY + SELECTION
# -------------------------------------------------
compatible_rows = []
rejections = []


for _, r in bat_df.iterrows():
    bat_v = n(r["nominal_voltage_v"])
    bat_nom_kwh = n(r["nominal_energy_kwh"])
    bat_usable_kwh = n(r["usable_energy_kwh"])
    max_parallel_units = int(n(r["max_parallel_units"], MAX_SEARCH_UNITS))


    if bat_v <= 0 or bat_nom_kwh <= 0:
        rejections.append({**r.to_dict(), "reasons": ["invalid_voltage_or_nominal_energy"]})
        continue


    ok, voltage_quality = voltage_compatible(inv_battery_voltage_v, bat_v, r["notes"])


    if not ok:
        rejections.append({**r.to_dict(), "reasons": [voltage_quality]})
        continue


    selection_energy_kwh = bat_nom_kwh


    # Single-unit diagnostic
    single_delivered_kwh = selection_energy_kwh
    single_shortfall_kwh = max(0.0, battery_balanced_kwh - single_delivered_kwh)
    single_excess_kwh = max(0.0, single_delivered_kwh - battery_balanced_kwh)
    single_variance_kwh = abs(single_delivered_kwh - battery_balanced_kwh)
    single_shortfall_pct = single_shortfall_kwh / battery_balanced_kwh * 100.0 if battery_balanced_kwh > 0 else 0.0
    single_variance_pct = single_variance_kwh / battery_balanced_kwh * 100.0 if battery_balanced_kwh > 0 else 0.0
    single_unit_within_tolerance = single_shortfall_pct <= SINGLE_UNIT_SHORTFALL_TOLERANCE_PCT


    unit_options = []


    for units in range(1, max_parallel_units + 1):
        delivered_nominal_kwh = units * selection_energy_kwh
        delivered_usable_kwh = units * bat_usable_kwh


        shortfall_kwh = max(0.0, battery_balanced_kwh - delivered_nominal_kwh)
        excess_kwh = max(0.0, delivered_nominal_kwh - battery_balanced_kwh)
        variance_kwh = abs(delivered_nominal_kwh - battery_balanced_kwh)


        shortfall_pct = shortfall_kwh / battery_balanced_kwh * 100.0 if battery_balanced_kwh > 0 else 0.0
        excess_pct = excess_kwh / battery_balanced_kwh * 100.0 if battery_balanced_kwh > 0 else 0.0
        variance_pct = variance_kwh / battery_balanced_kwh * 100.0 if battery_balanced_kwh > 0 else 0.0


        meets_bal_nominal = delivered_nominal_kwh >= battery_balanced_kwh
        within_tolerance = shortfall_pct <= SINGLE_UNIT_SHORTFALL_TOLERANCE_PCT


        score = 0
        score += 1000 if meets_bal_nominal else 0
        score += 700 if units == 1 and within_tolerance else 0
        score -= variance_pct * 20
        score -= excess_pct * 5
        score -= units * 15


        unit_options.append({
            "units": units,
            "delivered_nominal_kwh": delivered_nominal_kwh,
            "delivered_usable_kwh": delivered_usable_kwh,
            "shortfall_kwh": shortfall_kwh,
            "excess_kwh": excess_kwh,
            "variance_kwh": variance_kwh,
            "shortfall_pct": shortfall_pct,
            "excess_pct": excess_pct,
            "variance_pct": variance_pct,
            "meets_bal_nominal": meets_bal_nominal,
            "within_tolerance": within_tolerance,
            "score": score,
        })


    if single_unit_within_tolerance:
        selected_option = unit_options[0]
        selection_mode = "single_unit_within_balanced_tolerance"
    else:
        qualifying = [x for x in unit_options if x["meets_bal_nominal"]]
        if qualifying:
            selected_option = sorted(qualifying, key=lambda x: (x["variance_pct"], x["excess_pct"], x["units"]))[0]
            selection_mode = "nearest_nominal_meeting_balanced"
        else:
            selected_option = sorted(unit_options, key=lambda x: (x["variance_pct"], x["shortfall_pct"], x["units"]))[0]
            selection_mode = "nearest_nominal_even_if_short"


    candidate_score = selected_option["score"]
    if selected_option["units"] == 1:
        candidate_score += 500
    if voltage_quality == "exact_nominal_voltage_match":
        candidate_score += 30
    elif voltage_quality == "51_2v_battery_on_48v_inverter_class":
        candidate_score += 25
    else:
        candidate_score += 10


    # Ceil reference only
    ceil_reference_units_min = math.ceil(battery_minimum_kwh / selection_energy_kwh) if battery_minimum_kwh > 0 else 0
    ceil_reference_units_bal = math.ceil(battery_balanced_kwh / selection_energy_kwh) if battery_balanced_kwh > 0 else 0
    ceil_reference_units_fut = math.ceil(battery_future_kwh / selection_energy_kwh) if battery_future_kwh > 0 else 0


    ceil_reference_delivered_min_kwh = ceil_reference_units_min * selection_energy_kwh
    ceil_reference_delivered_balanced_kwh = ceil_reference_units_bal * selection_energy_kwh
    ceil_reference_delivered_future_kwh = ceil_reference_units_fut * selection_energy_kwh


    compatible_rows.append({
        **r.to_dict(),


        "selection_energy_basis": "nominal_energy_kwh",
        "selection_energy_kwh": selection_energy_kwh,
        "usable_energy_reference_kwh": bat_usable_kwh,
        "dc_mode": "parallel_only_low_voltage",
        "voltage_match_quality": voltage_quality,
        "selection_mode": selection_mode,


        # Clean selected design fields — downstream should use these
        "selected_units_bal": int(selected_option["units"]),
        "selected_delivered_balanced_kwh": selected_option["delivered_nominal_kwh"],
        "selected_delivered_usable_kwh": selected_option["delivered_usable_kwh"],
        "selected_shortfall_kwh": selected_option["shortfall_kwh"],
        "selected_excess_kwh": selected_option["excess_kwh"],
        "selected_variance_kwh": selected_option["variance_kwh"],
        "selected_shortfall_pct": selected_option["shortfall_pct"],
        "selected_excess_pct": selected_option["excess_pct"],
        "selected_variance_pct": selected_option["variance_pct"],
        "selected_meets_bal_nominal": int(selected_option["meets_bal_nominal"]),
        "selected_within_tolerance": int(selected_option["within_tolerance"]),


        # Explicit reference-only fields — do not use downstream for final sizing
        "ceil_reference_units_min": ceil_reference_units_min,
        "ceil_reference_units_bal": ceil_reference_units_bal,
        "ceil_reference_units_fut": ceil_reference_units_fut,
        "ceil_reference_delivered_min_kwh": ceil_reference_delivered_min_kwh,
        "ceil_reference_delivered_balanced_kwh": ceil_reference_delivered_balanced_kwh,
        "ceil_reference_delivered_future_kwh": ceil_reference_delivered_future_kwh,
        "ceil_reference_variance_balanced_kwh": abs(ceil_reference_delivered_balanced_kwh - battery_balanced_kwh),
        "ceil_reference_excess_balanced_kwh": max(0.0, ceil_reference_delivered_balanced_kwh - battery_balanced_kwh),
        "ceil_reference_shortfall_balanced_kwh": max(0.0, battery_balanced_kwh - ceil_reference_delivered_balanced_kwh),
        "ceil_reference_meets_min": int(ceil_reference_delivered_min_kwh >= battery_minimum_kwh),
        "ceil_reference_meets_bal": int(ceil_reference_delivered_balanced_kwh >= battery_balanced_kwh),
        "ceil_reference_meets_fut": int(ceil_reference_delivered_future_kwh >= battery_future_kwh),


        # Single-unit diagnostic
        "single_unit_delivered_nominal_kwh": single_delivered_kwh,
        "single_unit_shortfall_kwh": single_shortfall_kwh,
        "single_unit_excess_kwh": single_excess_kwh,
        "single_unit_variance_kwh": single_variance_kwh,
        "single_unit_shortfall_pct": single_shortfall_pct,
        "single_unit_variance_pct": single_variance_pct,
        "single_unit_within_tolerance": int(single_unit_within_tolerance),


        "unit_options": unit_options,
        "score": candidate_score,
    })


if compatible_rows:
    compatible_df = pd.DataFrame(compatible_rows).sort_values(
        [
            "single_unit_within_tolerance",
            "selected_units_bal",
            "selected_variance_pct",
            "selected_excess_pct",
            "score",
        ],
        ascending=[False, True, True, True, False]
    ).reset_index(drop=True)
else:
    compatible_df = pd.DataFrame()


print("Compatible battery candidates:", len(compatible_df))
if len(compatible_df):
    print(compatible_df[[
        "identifier",
        "nominal_voltage_v",
        "nominal_energy_kwh",
        "usable_energy_kwh",
        "selection_mode",
        "selected_units_bal",
        "selected_delivered_balanced_kwh",
        "selected_variance_pct",
        "single_unit_within_tolerance",
        "score"
    ]].to_string(index=False))
else:
    print("No compatible battery candidates found.")




# -------------------------------------------------
# SELECT + SAVE
# -------------------------------------------------
if len(compatible_df):
    selected_battery = compatible_df.iloc[0].to_dict()
    shortlist = compatible_df.head(10).to_dict(orient="records")
    selection_status = "selected"
else:
    selected_battery = {}
    shortlist = []
    selection_status = "no_candidate_found"


battery_state_path = os.path.join(design_dir, "design_state_step_02_battery.json")
battery_report_md_path = os.path.join(design_dir, "design_step_02_battery_report.md")
battery_debug_path = os.path.join(design_dir, "battery_candidates_debug.json")


battery_state = {
    "meta": {
        "generated_at": datetime.now().isoformat(),
        "submission_key": latest_key,
        "project_root": project_root,
        "source_inverter_state": inverter_state_path,
        "source_load_state": load_state_path,
        "source_battery_schema": BATTERY_PATH,
    },
    "selection_inputs": {
        "selected_inverter_identifier": inv_identifier,
        "inverter_battery_voltage_nominal_v": inv_battery_voltage_v,
        "inverter_battery_architecture": inv_battery_architecture,
        "battery_minimum_kwh": battery_minimum_kwh,
        "battery_balanced_kwh": battery_balanced_kwh,
        "battery_future_kwh": battery_future_kwh,
        "selection_basis": "nominal_energy_kwh",
        "battery_usable_dod_reference": BATTERY_USABLE_DOD,
        "single_unit_shortfall_tolerance_pct": SINGLE_UNIT_SHORTFALL_TOLERANCE_PCT,
    },
    "selection_status": selection_status,
    "selected_battery": selected_battery,
    "shortlist": shortlist,
    "rejections_preview": rejections[:100],
    "selection_reasons": [
        "Battery modules evaluated on nominal energy for selection logic.",
        "Space is prioritized: one battery unit is considered first.",
        "If a single-unit option is within tolerance of the balanced target, it is preferred.",
        "Ceil-reference fields are retained only as diagnostics and should not drive downstream sizing.",
        "Downstream blocks should use selected_* fields only.",
        "48V-class low-voltage inverter compatibility accepts 51.2V low-voltage battery modules.",
    ],
}


save_json(battery_state_path, battery_state)


save_json(battery_debug_path, {
    "meta": {
        "generated_at": datetime.now().isoformat(),
        "submission_key": latest_key,
        "project_root": project_root,
    },
    "inverter_context": {
        "identifier": inv_identifier,
        "battery_voltage_nominal_v": inv_battery_voltage_v,
        "battery_architecture": inv_battery_architecture,
    },
    "battery_targets": {
        "minimum_kwh": battery_minimum_kwh,
        "balanced_kwh": battery_balanced_kwh,
        "future_kwh": battery_future_kwh,
        "selection_basis": "nominal_energy_kwh",
        "single_unit_shortfall_tolerance_pct": SINGLE_UNIT_SHORTFALL_TOLERANCE_PCT,
    },
    "compatible_count": int(len(compatible_df)),
    "compatible_preview": compatible_df.head(20).to_dict(orient="records") if len(compatible_df) else [],
    "rejections_preview": rejections[:100],
})


if selected_battery:
    report_text = f"""# TENET Design Step 02 — Battery Selection Report


## Selected Battery
- Identifier: **{selected_battery.get('identifier','')}**
- Nominal Voltage: **{n(selected_battery.get('nominal_voltage_v'),0):.2f} V**
- Nominal Energy per Unit: **{n(selected_battery.get('nominal_energy_kwh'),0):.4f} kWh**
- Usable Energy per Unit Reference at {BATTERY_USABLE_DOD:.0%} DoD: **{n(selected_battery.get('usable_energy_kwh'),0):.4f} kWh**
- Selection Mode: **{selected_battery.get('selection_mode','')}**
- Selected Units for Balanced Target: **{int(n(selected_battery.get('selected_units_bal'),0))}**
- Selected Delivered Balanced Nominal Energy: **{n(selected_battery.get('selected_delivered_balanced_kwh'),0):.4f} kWh**
- Selected Delivered Usable Energy Reference: **{n(selected_battery.get('selected_delivered_usable_kwh'),0):.4f} kWh**
- Selected Variance from Balanced Target: **{n(selected_battery.get('selected_variance_pct'),0):.2f}%**
- Selected Shortfall from Balanced Target: **{n(selected_battery.get('selected_shortfall_pct'),0):.2f}%**
- Single Unit Within Tolerance: **{selected_battery.get('single_unit_within_tolerance')}**
- Voltage Match Quality: **{selected_battery.get('voltage_match_quality','')}**


## Reference-only Fields
The `ceil_reference_*` fields are retained only for diagnostics. Downstream blocks should use the `selected_*` fields.


## Targets
- Battery Minimum: **{battery_minimum_kwh:.4f} kWh**
- Battery Balanced: **{battery_balanced_kwh:.4f} kWh**
- Battery Future: **{battery_future_kwh:.4f} kWh**


## Canonical Handoff
- JSON State: `{battery_state_path}`
- Debug File: `{battery_debug_path}`
"""
else:
    report_text = f"""# TENET Design Step 02 — Battery Selection Report


No compatible battery candidate found.


## Canonical Handoff
- JSON State: `{battery_state_path}`
- Debug File: `{battery_debug_path}`
"""


with open(battery_report_md_path, "w", encoding="utf-8") as f:
    f.write(report_text)


print()
print("----- BLOCK 4 COMPLETE -----")
print("Selection status:", selection_status)
print("JSON handoff:", battery_state_path)
print("Markdown report:", battery_report_md_path)
print("Debug:", battery_debug_path)


if selected_battery:
    print()
    print("Selected battery:", selected_battery.get("identifier"))
    print("Selection mode:", selected_battery.get("selection_mode"))
    print("Nominal energy per unit (kWh):", round(n(selected_battery.get("nominal_energy_kwh"), 0), 4))
    print("Selected units:", int(n(selected_battery.get("selected_units_bal"), 0)))
    print("Delivered nominal balanced kWh:", round(n(selected_battery.get("selected_delivered_balanced_kwh"), 0), 4))
    print("Variance %:", round(n(selected_battery.get("selected_variance_pct"), 0), 2))

```

```python
# =========================
# BLOCK 5 REVISION
# SUB-BLOCK 1/3
# Intent-led kWp sizing logic only
# CLEANED FOR NEW BATTERY HANDOFF
# =========================


import os
import json
import math
from datetime import datetime
from typing import Any


import numpy as np
import pandas as pd


# -------------------------------------------------
# CONFIG
# -------------------------------------------------
PROJECTS_ROOT = "/content/drive/My Drive/2026/tenet/SSOT/RFQI v2/submissions/"
REGISTRY_PATH = os.path.join(PROJECTS_ROOT, "submission_registry.json")


# Intent-led design assumptions
DEFAULT_PSH = 5.0
SYSTEM_DERATE = 0.78


DAY_FRACTION = 9.0 / 24.0
NIGHT_FRACTION = 15.0 / 24.0


BATTERY_READY_WINDOW_HOURS = 3.5
DESIGN_MARGIN_FACTOR = 1.10


# -------------------------------------------------
# HELPERS
# -------------------------------------------------
def n(v: Any, default: float = 0.0) -> float:
    if v is None:
        return default
    try:
        if isinstance(v, float) and math.isnan(v):
            return default
    except Exception:
        pass
    try:
        if pd.isna(v):
            return default
    except Exception:
        pass
    try:
        if isinstance(v, str):
            cleaned = (
                v.replace(",", "")
                 .replace("kWh", "")
                 .replace("kW", "")
                 .replace("W", "")
                 .replace("V", "")
                 .replace("A", "")
                 .replace("%", "")
                 .strip()
            )
            return float(cleaned)
        return float(v)
    except Exception:
        return default




def s(v: Any) -> str:
    if v is None:
        return ""
    try:
        if pd.isna(v):
            return ""
    except Exception:
        pass
    return str(v).strip()




def norm_text(x: Any) -> str:
    x = s(x).lower()
    x = x.replace("&", "and")
    x = x.replace("-", " ")
    x = x.replace("/", " ")
    x = x.replace("_", " ")
    x = " ".join(x.split())
    return x




def make_json_safe(obj):
    if isinstance(obj, dict):
        return {str(k): make_json_safe(v) for k, v in obj.items()}
    if isinstance(obj, (list, tuple, set)):
        return [make_json_safe(v) for v in obj]
    if isinstance(obj, (pd.Timestamp, datetime)):
        return obj.isoformat()
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.floating):
        return float(obj)
    if isinstance(obj, np.bool_):
        return bool(obj)
    if isinstance(obj, np.ndarray):
        return [make_json_safe(v) for v in obj.tolist()]
    try:
        if pd.isna(obj):
            return None
    except Exception:
        pass
    if isinstance(obj, (str, int, float, bool)) or obj is None:
        return obj
    return str(obj)




def save_json(path: str, data: dict):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(make_json_safe(data), f, indent=4, ensure_ascii=False)




def load_json(path: str):
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing file: {path}")
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)




def resolve_current_submission(projects_root: str, registry_path: str):
    if not os.path.exists(projects_root):
        raise FileNotFoundError(f"Submissions root not found: {projects_root}")


    if os.path.exists(registry_path):
        registry = load_json(registry_path)
        submissions = registry.get("submissions", {})
        if submissions:
            latest_key = max(submissions.keys(), key=lambda k: submissions[k].get("last_seen_at", ""))
            return {
                "source": "registry",
                "latest_key": latest_key,
                "project_root": submissions[latest_key]["project_root"],
            }


    candidates = []
    for name in os.listdir(projects_root):
        full = os.path.join(projects_root, name)
        if os.path.isdir(full):
            load_state = os.path.join(full, "01_load", "load_state.json")
            inv_state = os.path.join(full, "02_design", "design_state_step_01_inverter.json")
            bat_state = os.path.join(full, "02_design", "design_state_step_02_battery.json")
            if os.path.exists(load_state) and os.path.exists(inv_state) and os.path.exists(bat_state):
                candidates.append((os.path.getmtime(full), name, full))


    if not candidates:
        raise FileNotFoundError("No valid submission folders found.")


    candidates.sort(reverse=True)
    _, latest_key, project_root = candidates[0]


    return {
        "source": "folder_scan",
        "latest_key": latest_key,
        "project_root": project_root,
    }


# -------------------------------------------------
# FIND CURRENT SUBMISSION
# -------------------------------------------------
resolved = resolve_current_submission(PROJECTS_ROOT, REGISTRY_PATH)


latest_key = resolved["latest_key"]
project_root = resolved["project_root"]
resolution_source = resolved["source"]


design_dir = os.path.join(project_root, "02_design")
load_dir = os.path.join(project_root, "01_load")
os.makedirs(design_dir, exist_ok=True)


load_state_path = os.path.join(load_dir, "load_state.json")
inverter_state_path = os.path.join(design_dir, "design_state_step_01_inverter.json")
battery_state_path = os.path.join(design_dir, "design_state_step_02_battery.json")


load_state = load_json(load_state_path)
inverter_state = load_json(inverter_state_path)
battery_state = load_json(battery_state_path)


# -------------------------------------------------
# READ INPUTS
# -------------------------------------------------
energy_breakdown = load_state.get("energy_breakdown", {})
selected_inverter = inverter_state.get("selected_inverter", {})
selected_battery = battery_state.get("selected_battery", {})


inv_identifier = s(selected_inverter.get("identifier"))
inv_rated_power_kw = n(selected_inverter.get("rated_power_kw"), 0.0)
inv_mppt_count = int(n(selected_inverter.get("mppt_count"), 1))


inv_max_pv_input_power_w = n(selected_inverter.get("max_pv_input_power_w"), 0.0)
inv_max_pv_input_voltage_v = n(selected_inverter.get("max_pv_input_voltage_v"), 0.0)
inv_mppt_voltage_min_v = n(selected_inverter.get("mppt_voltage_min_v"), 0.0)
inv_mppt_voltage_max_v = n(selected_inverter.get("mppt_voltage_max_v"), 0.0)
inv_max_pv_input_current_a = n(selected_inverter.get("max_pv_input_current_a"), 0.0)


battery_identifier = s(selected_battery.get("identifier"))
battery_selection_mode = s(selected_battery.get("selection_mode"))


selected_battery_units_bal = int(n(selected_battery.get("selected_units_bal"), 0))
selected_battery_nominal_kwh = n(selected_battery.get("selected_delivered_balanced_kwh"), 0.0)
selected_battery_usable_kwh = n(selected_battery.get("selected_delivered_usable_kwh"), 0.0)


if selected_battery_units_bal <= 0:
    raise ValueError("Battery selected_units_bal missing or invalid. Rerun cleaned Block 4.")


if selected_battery_nominal_kwh <= 0:
    raise ValueError("Battery selected_delivered_balanced_kwh missing or invalid. Rerun cleaned Block 4.")


if selected_battery_usable_kwh <= 0:
    selected_battery_usable_kwh = selected_battery_nominal_kwh * 0.90


# Total daily duty from load block
e_total_daily_kwh = n(energy_breakdown.get("e_required_kwh"), 0.0)


if e_total_daily_kwh <= 0:
    raise ValueError("Total daily energy requirement missing or invalid from load_state.energy_breakdown.e_required_kwh")


# Day/night split
e_day_load_kwh = e_total_daily_kwh * DAY_FRACTION
e_night_load_kwh = e_total_daily_kwh * NIGHT_FRACTION


# -------------------------------------------------
# INTENT-LED PV ENERGY DUTY
# -------------------------------------------------
# Product selection uses battery nominal kWh.
# PV recharge duty uses usable battery kWh reference because that is the practical night-duty energy.
battery_recharge_duty_kwh = min(
    e_night_load_kwh,
    selected_battery_usable_kwh
)


unserved_night_duty_kwh = max(
    0.0,
    e_night_load_kwh - selected_battery_usable_kwh
)


e_pv_target_kwh = e_day_load_kwh + battery_recharge_duty_kwh


# -------------------------------------------------
# kWp SIZING LOGIC
# -------------------------------------------------
pv_kwp_raw = (
    e_pv_target_kwh / (DEFAULT_PSH * SYSTEM_DERATE)
    if DEFAULT_PSH > 0 and SYSTEM_DERATE > 0 else 0.0
)


pv_kwp_target_after_margin = pv_kwp_raw * DESIGN_MARGIN_FACTOR


pv_kwp_midday_charge_requirement = (
    battery_recharge_duty_kwh / (BATTERY_READY_WINDOW_HOURS * SYSTEM_DERATE)
    if BATTERY_READY_WINDOW_HOURS > 0 and SYSTEM_DERATE > 0 else 0.0
)


pv_kwp_final_target = max(
    pv_kwp_target_after_margin,
    pv_kwp_midday_charge_requirement
)


estimated_surplus_trigger_note = (
    "Any generation above the daytime-load + selected usable battery recharge duty "
    "is treated as a recruitment window for shiftable loads, not the primary sizing basis."
)


battery_note = (
    "Battery product selection used nominal kWh, but PV recharge duty uses selected usable kWh "
    "because DoD limits practical night-duty energy."
)


# -------------------------------------------------
# PRINT
# -------------------------------------------------
print("Submission resolution source:", resolution_source)
print("Current submission key:", latest_key)
print("Project root:", project_root)
print()
print("Selected inverter:", inv_identifier)
print("Inverter rated power (kW):", inv_rated_power_kw)
print("MPPT count:", inv_mppt_count)
print("Max PV input power (W):", inv_max_pv_input_power_w)
print("Max PV input voltage (V):", inv_max_pv_input_voltage_v)
print("MPPT voltage min/max (V):", inv_mppt_voltage_min_v, "/", inv_mppt_voltage_max_v)
print("Max PV input current (A):", inv_max_pv_input_current_a)
print()
print("Selected battery:", battery_identifier)
print("Battery selection mode:", battery_selection_mode)
print("Selected battery units:", selected_battery_units_bal)
print("Selected battery nominal kWh:", round(selected_battery_nominal_kwh, 4))
print("Selected battery usable kWh reference:", round(selected_battery_usable_kwh, 4))
print()
print("Total daily load (kWh):", round(e_total_daily_kwh, 4))
print("Estimated daytime load (kWh):", round(e_day_load_kwh, 4))
print("Estimated night load (kWh):", round(e_night_load_kwh, 4))
print("Battery recharge duty used by PV block (kWh):", round(battery_recharge_duty_kwh, 4))
print("Unserved night duty beyond selected usable battery (kWh):", round(unserved_night_duty_kwh, 4))
print()
print("PV target energy (kWh/day):", round(e_pv_target_kwh, 4))
print("Raw PV size before margin (kWp):", round(pv_kwp_raw, 4))
print("Design margin factor:", DESIGN_MARGIN_FACTOR)
print("PV target after margin (kWp):", round(pv_kwp_target_after_margin, 4))
print("Midday battery readiness requirement (kWp):", round(pv_kwp_midday_charge_requirement, 4))
print("Final PV target size (kWp):", round(pv_kwp_final_target, 4))
print()


# -------------------------------------------------
# SAVE INTERMEDIATE STATE
# -------------------------------------------------
pv_stage_01_path = os.path.join(design_dir, "pv_stage_01_target_logic.json")


save_json(pv_stage_01_path, {
    "meta": {
        "generated_at": datetime.now().isoformat(),
        "submission_key": latest_key,
        "project_root": project_root,
        "submission_resolution_source": resolution_source,
        "source_load_state": load_state_path,
        "source_inverter_state": inverter_state_path,
        "source_battery_state": battery_state_path,
    },
    "inputs": {
        "selected_inverter_identifier": inv_identifier,
        "selected_battery_identifier": battery_identifier,
        "battery_selection_mode": battery_selection_mode,
        "selected_battery_units_bal": selected_battery_units_bal,


        "inverter_rated_power_kw": inv_rated_power_kw,
        "inverter_mppt_count": inv_mppt_count,
        "inverter_max_pv_input_power_w": inv_max_pv_input_power_w,
        "inverter_max_pv_input_voltage_v": inv_max_pv_input_voltage_v,
        "inverter_mppt_voltage_min_v": inv_mppt_voltage_min_v,
        "inverter_mppt_voltage_max_v": inv_mppt_voltage_max_v,
        "inverter_max_pv_input_current_a": inv_max_pv_input_current_a,


        "e_total_daily_kwh": e_total_daily_kwh,
        "e_day_load_kwh": e_day_load_kwh,
        "e_night_load_kwh": e_night_load_kwh,


        "selected_battery_nominal_kwh": selected_battery_nominal_kwh,
        "selected_battery_usable_kwh": selected_battery_usable_kwh,
        "battery_recharge_duty_kwh": battery_recharge_duty_kwh,
        "unserved_night_duty_kwh": unserved_night_duty_kwh,


        "psh": DEFAULT_PSH,
        "system_derate": SYSTEM_DERATE,
        "battery_ready_window_hours": BATTERY_READY_WINDOW_HOURS,
        "design_margin_factor": DESIGN_MARGIN_FACTOR,
    },
    "results": {
        "e_pv_target_kwh": e_pv_target_kwh,
        "pv_kwp_raw": pv_kwp_raw,
        "pv_kwp_target_after_margin": pv_kwp_target_after_margin,
        "pv_kwp_midday_charge_requirement": pv_kwp_midday_charge_requirement,
        "pv_kwp_final_target": pv_kwp_final_target,
        "battery_note": battery_note,
        "surplus_note": estimated_surplus_trigger_note,
    },
    "downstream_instruction": {
        "pv_product_selection_should_use": "pv_kwp_final_target",
        "battery_fields_to_use": [
            "selected_battery_nominal_kwh",
            "selected_battery_usable_kwh",
            "selected_battery_units_bal",
            "battery_recharge_duty_kwh"
        ],
        "do_not_use_legacy_battery_fields": [
            "units_bal",
            "delivered_balanced_kwh",
            "ceil_reference_units_bal",
            "ceil_reference_delivered_balanced_kwh"
        ]
    }
})


print("Saved intermediate target logic to:")
print(pv_stage_01_path)

```

```python
# =========================
# BLOCK 5 REVISION
# SUB-BLOCK 2/3
# PV PRODUCT SELECTION + ARRAY DESIGN
# FIXED WITH STRONG PV COLUMN AUTO-DETECTION
# =========================


import os
import json
import math
import re
from datetime import datetime
from typing import Any


import numpy as np
import pandas as pd


# -------------------------------------------------
# CONFIG
# -------------------------------------------------
PROJECTS_ROOT = "/content/drive/My Drive/2026/tenet/SSOT/RFQI v2/submissions/"
REGISTRY_PATH = os.path.join(PROJECTS_ROOT, "submission_registry.json")


PV_DIR = "/content/drive/My Drive/2026/tenet/SSOT/RFQI v2/component_files/"
PV_FILE = "TENET_PV_unified_module_schema_v2.xlsx"
PV_PATH = os.path.join(PV_DIR, PV_FILE)
PV_SHEET = 0


DEFAULT_VOC_COLD_FACTOR = 1.10
DEFAULT_VMP_HOT_FACTOR = 0.95


DEFAULT_INV_MAX_VDC = 500.0
DEFAULT_INV_MPPT_LOW = 60.0
DEFAULT_INV_MPPT_HIGH = 400.0
DEFAULT_INV_MAX_INPUT_CURRENT_A = 187.5


PV_OVERSIZE_LIMIT_RATIO = 1.50
ALLOWED_DC_AC_RATIO = 1.50


# -------------------------------------------------
# HELPERS
# -------------------------------------------------
def n(v: Any, default: float = 0.0) -> float:
    if v is None:
        return default
    try:
        if isinstance(v, float) and math.isnan(v):
            return default
    except Exception:
        pass
    try:
        if pd.isna(v):
            return default
    except Exception:
        pass
    try:
        if isinstance(v, str):
            cleaned = (
                v.replace(",", "")
                 .replace("kWh", "")
                 .replace("kW", "")
                 .replace("Wp", "")
                 .replace("W", "")
                 .replace("V", "")
                 .replace("A", "")
                 .replace("%", "")
                 .replace("m²", "")
                 .replace("m2", "")
                 .strip()
            )
            return float(cleaned)
        return float(v)
    except Exception:
        return default


def s(v: Any) -> str:
    if v is None:
        return ""
    try:
        if pd.isna(v):
            return ""
    except Exception:
        pass
    return str(v).strip()


def normalize_col(c: Any) -> str:
    c = s(c).lower()
    for ch in [" ", "_", "-", "/", "\\", "(", ")", "[", "]", ".", ",", ":", ";", "%"]:
        c = c.replace(ch, "")
    return c


def make_json_safe(obj):
    if isinstance(obj, dict):
        return {str(k): make_json_safe(v) for k, v in obj.items()}
    if isinstance(obj, (list, tuple, set)):
        return [make_json_safe(v) for v in obj]
    if isinstance(obj, (pd.Timestamp, datetime)):
        return obj.isoformat()
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.floating):
        return float(obj)
    if isinstance(obj, np.bool_):
        return bool(obj)
    try:
        if pd.isna(obj):
            return None
    except Exception:
        pass
    return obj if isinstance(obj, (str, int, float, bool)) or obj is None else str(obj)


def save_json(path: str, data: dict):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(make_json_safe(data), f, indent=4, ensure_ascii=False)


def load_json(path: str):
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing file: {path}")
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def resolve_current_submission(projects_root: str, registry_path: str):
    if os.path.exists(registry_path):
        registry = load_json(registry_path)
        submissions = registry.get("submissions", {})
        if submissions:
            latest_key = max(submissions.keys(), key=lambda k: submissions[k].get("last_seen_at", ""))
            return latest_key, submissions[latest_key]["project_root"], "registry"


    candidates = []
    for name in os.listdir(projects_root):
        full = os.path.join(projects_root, name)
        if os.path.isdir(full):
            stage1 = os.path.join(full, "02_design", "pv_stage_01_target_logic.json")
            inv = os.path.join(full, "02_design", "design_state_step_01_inverter.json")
            if os.path.exists(stage1) and os.path.exists(inv):
                candidates.append((os.path.getmtime(full), name, full))


    if not candidates:
        raise FileNotFoundError("No valid submission folders found for PV selection.")


    candidates.sort(reverse=True)
    return candidates[0][1], candidates[0][2], "folder_scan"


def find_column(columns, exact_aliases=None, contains_all=None, contains_any=None, exclude_any=None):
    exact_aliases = exact_aliases or []
    contains_all = contains_all or []
    contains_any = contains_any or []
    exclude_any = exclude_any or []


    col_map = {col: normalize_col(col) for col in columns}


    for col, cn in col_map.items():
        for alias in exact_aliases:
            if cn == normalize_col(alias):
                return col


    for col, cn in col_map.items():
        if exclude_any and any(normalize_col(x) in cn for x in exclude_any):
            continue
        if contains_all and not all(normalize_col(x) in cn for x in contains_all):
            continue
        if contains_any and not any(normalize_col(x) in cn for x in contains_any):
            continue
        return col


    return None


def parse_power_from_model(model_text):
    txt = s(model_text)
    match = re.search(r"(\d{3,4})\s*w", txt.lower())
    if match:
        return float(match.group(1))
    return 0.0


# -------------------------------------------------
# LOAD CONTEXT
# -------------------------------------------------
latest_key, project_root, resolution_source = resolve_current_submission(PROJECTS_ROOT, REGISTRY_PATH)


design_dir = os.path.join(project_root, "02_design")
pv_stage_01_path = os.path.join(design_dir, "pv_stage_01_target_logic.json")
inverter_state_path = os.path.join(design_dir, "design_state_step_01_inverter.json")


pv_stage_01 = load_json(pv_stage_01_path)
inverter_state = load_json(inverter_state_path)


stage1_inputs = pv_stage_01.get("inputs", {})
stage1_results = pv_stage_01.get("results", {})
selected_inverter = inverter_state.get("selected_inverter", {})


inv_identifier = s(stage1_inputs.get("selected_inverter_identifier"))
inv_rated_power_kw = n(stage1_inputs.get("inverter_rated_power_kw"), n(selected_inverter.get("rated_power_kw"), 0.0))


pv_kwp_raw = n(stage1_results.get("pv_kwp_raw"), 0.0)
pv_kwp_final_target = n(stage1_results.get("pv_kwp_final_target"), 0.0)


inv_max_vdc = n(selected_inverter.get("max_pv_input_voltage_v"), DEFAULT_INV_MAX_VDC)
inv_mppt_low = n(selected_inverter.get("mppt_voltage_min_v"), DEFAULT_INV_MPPT_LOW)
inv_mppt_high = n(selected_inverter.get("mppt_voltage_max_v"), DEFAULT_INV_MPPT_HIGH)
inv_max_input_current_a = n(selected_inverter.get("max_input_current_a"), DEFAULT_INV_MAX_INPUT_CURRENT_A)
inv_max_pv_input_power_w = n(selected_inverter.get("max_pv_input_power_w"), 0.0)


if inv_max_pv_input_power_w <= 0 and inv_rated_power_kw > 0:
    inv_max_pv_input_power_w = inv_rated_power_kw * 1000.0 * ALLOWED_DC_AC_RATIO


# -------------------------------------------------
# LOAD PV SCHEMA
# -------------------------------------------------
pv_raw_df = pd.read_excel(PV_PATH, sheet_name=PV_SHEET)
cols = list(pv_raw_df.columns)


# -------------------------------------------------
# AUTO-DETECT COLUMNS
# -------------------------------------------------
brand_col = find_column(cols, exact_aliases=["brand"])
model_col = find_column(cols, exact_aliases=["model", "module_model", "module_name"])
series_col = find_column(cols, exact_aliases=["series"])


p_col = find_column(
    cols,
    exact_aliases=["nameplate_power_w", "nameplate power w", "pmax_w", "rated_power_w", "module_power_w"],
    contains_all=["power"],
    contains_any=["nameplate", "pmax", "rated"],
    exclude_any=["tolerance", "warranty"]
)


voc_col = find_column(
    cols,
    exact_aliases=["open_circuit_voltage_v", "open circuit voltage voc", "open circuit voltage (voc)", "voc"],
    contains_all=["voc"],
    contains_any=["voltage", "circuit", "open"]
)


isc_col = find_column(
    cols,
    exact_aliases=["short_circuit_current_a", "short circuit current isc", "short circuit current (isc)", "isc"],
    contains_all=["isc"],
    contains_any=["current", "circuit", "short"]
)


vmp_col = find_column(
    cols,
    exact_aliases=[
        "maximum_power_point_voltage_v",
        "maximum power point voltage vmp",
        "maximum power point voltage (vmp)",
        "max_power_voltage_v",
        "mpp_voltage_v",
        "vmp"
    ],
    contains_all=["vmp"],
    contains_any=["voltage", "power", "point", "max"]
)


imp_col = find_column(
    cols,
    exact_aliases=[
        "maximum_power_point_current_a",
        "maximum power point current imp",
        "maximum power point current (imp)",
        "max_power_current_a",
        "mpp_current_a",
        "imp"
    ],
    contains_all=["imp"],
    contains_any=["current", "power", "point", "max"]
)


# fallback aliases if columns use long text without vmp/imp tokens
if vmp_col is None:
    vmp_col = find_column(cols, contains_all=["maximum", "power", "voltage"])
if vmp_col is None:
    vmp_col = find_column(cols, contains_all=["max", "power", "voltage"])


if imp_col is None:
    imp_col = find_column(cols, contains_all=["maximum", "power", "current"])
if imp_col is None:
    imp_col = find_column(cols, contains_all=["max", "power", "current"])


print("Detected PV columns:")
print("brand:", brand_col)
print("model:", model_col)
print("series:", series_col)
print("power:", p_col)
print("voc:", voc_col)
print("isc:", isc_col)
print("vmp:", vmp_col)
print("imp:", imp_col)
print()


# -------------------------------------------------
# NORMALIZE PV ROWS
# -------------------------------------------------
normalized_rows = []
rejections = []


for idx, r in pv_raw_df.iterrows():
    brand = s(r[brand_col]) if brand_col else ""
    model = s(r[model_col]) if model_col else ""
    series = s(r[series_col]) if series_col else ""


    identifier = " | ".join([x for x in [brand, model, series] if x]).strip()
    if not identifier:
        identifier = f"row_{idx}"


    p_w = n(r[p_col]) if p_col else 0.0
    if p_w <= 0:
        p_w = parse_power_from_model(model)


    voc = n(r[voc_col]) if voc_col else 0.0
    isc = n(r[isc_col]) if isc_col else 0.0
    vmp = n(r[vmp_col]) if vmp_col else 0.0
    imp = n(r[imp_col]) if imp_col else 0.0


    normalized_rows.append({
        "source_row_index": int(idx),
        "identifier": identifier,
        "brand": brand,
        "model": model,
        "series": series,
        "p_w": p_w,
        "voc": voc,
        "isc": isc,
        "vmp": vmp,
        "imp": imp,
    })


pv_df = pd.DataFrame(normalized_rows)


print("PV schema loaded:", PV_PATH)
print("Rows:", len(pv_raw_df))
print("Normalized preview:")
print(pv_df[["identifier", "p_w", "voc", "isc", "vmp", "imp"]].head(10).to_string(index=False))
print()


# -------------------------------------------------
# COMPATIBILITY SEARCH
# -------------------------------------------------
candidates = []


for _, r in pv_df.iterrows():
    p_w = n(r["p_w"])
    voc = n(r["voc"])
    isc = n(r["isc"])
    vmp = n(r["vmp"])
    imp = n(r["imp"])


    if p_w <= 0 or voc <= 0 or isc <= 0 or vmp <= 0 or imp <= 0:
        rejections.append({**r.to_dict(), "reasons": ["invalid_module_core_fields"]})
        continue


    module_kw = p_w / 1000.0


    max_series_by_vdc = int(math.floor(inv_max_vdc / (voc * DEFAULT_VOC_COLD_FACTOR)))
    min_series_by_mppt = int(math.ceil(inv_mppt_low / max(vmp * DEFAULT_VMP_HOT_FACTOR, 1e-9)))
    max_series_by_mppt = int(math.floor(inv_mppt_high / max(vmp * DEFAULT_VMP_HOT_FACTOR, 1e-9)))


    string_min = max(1, min_series_by_mppt)
    string_max = min(max_series_by_vdc, max_series_by_mppt)


    if string_max < string_min:
        rejections.append({
            **r.to_dict(),
            "reasons": ["no_valid_series_window"],
            "string_min": string_min,
            "string_max": string_max
        })
        continue


    for modules_per_string in range(string_min, string_max + 1):
        cold_string_voc = modules_per_string * voc * DEFAULT_VOC_COLD_FACTOR
        operating_string_vmp = modules_per_string * vmp * DEFAULT_VMP_HOT_FACTOR


        max_parallel_by_current = int(math.floor(inv_max_input_current_a / isc))
        max_parallel_by_power = int(math.floor(inv_max_pv_input_power_w / (modules_per_string * p_w)))


        max_parallel_strings = min(max_parallel_by_current, max_parallel_by_power)


        if max_parallel_strings <= 0:
            continue


        target_strings = max(1, int(round(pv_kwp_final_target / (modules_per_string * module_kw))))


        for num_strings in range(max(1, target_strings - 5), min(max_parallel_strings, target_strings + 5) + 1):
            total_modules = modules_per_string * num_strings
            total_kwp = total_modules * module_kw
            total_dc_power_w = total_modules * p_w
            array_isc = num_strings * isc
            array_imp = num_strings * imp


            if array_isc > inv_max_input_current_a:
                continue
            if total_dc_power_w > inv_max_pv_input_power_w:
                continue


            oversize_ratio = total_kwp / pv_kwp_final_target if pv_kwp_final_target > 0 else 999
            dc_ac_ratio = total_kwp / inv_rated_power_kw if inv_rated_power_kw > 0 else 999
            kwp_abs_error = abs(total_kwp - pv_kwp_final_target)


            if oversize_ratio > PV_OVERSIZE_LIMIT_RATIO:
                continue
            if dc_ac_ratio > ALLOWED_DC_AC_RATIO:
                continue


            candidates.append({
                **r.to_dict(),
                "module_kw": module_kw,
                "modules_per_string": modules_per_string,
                "num_strings": num_strings,
                "total_modules": total_modules,
                "total_kwp": total_kwp,
                "total_dc_power_w": total_dc_power_w,
                "kwp_abs_error": kwp_abs_error,
                "oversize_ratio": oversize_ratio,
                "dc_ac_ratio": dc_ac_ratio,
                "cold_string_voc": cold_string_voc,
                "operating_string_vmp": operating_string_vmp,
                "array_isc": array_isc,
                "array_imp": array_imp,
            })


# -------------------------------------------------
# SELECT
# -------------------------------------------------
if candidates:
    cand_df = pd.DataFrame(candidates).sort_values(
        ["kwp_abs_error", "oversize_ratio", "total_modules", "num_strings"],
        ascending=[True, True, True, True]
    ).reset_index(drop=True)


    selected_pv = cand_df.iloc[0].to_dict()
    shortlist = cand_df.head(10).to_dict(orient="records")
else:
    cand_df = pd.DataFrame()
    selected_pv = {}
    shortlist = []


# -------------------------------------------------
# SAVE OUTPUTS
# -------------------------------------------------
pv_state_path = os.path.join(design_dir, "design_state_step_03_pv.json")
pv_report_md_path = os.path.join(design_dir, "design_step_03_pv_report.md")
pv_debug_path = os.path.join(design_dir, "pv_candidates_debug.json")


save_json(pv_debug_path, {
    "candidate_count": int(len(cand_df)),
    "detected_columns": {
        "brand": brand_col,
        "model": model_col,
        "series": series_col,
        "power": p_col,
        "voc": voc_col,
        "isc": isc_col,
        "vmp": vmp_col,
        "imp": imp_col,
    },
    "normalized_rows_preview": pv_df.head(20).to_dict(orient="records"),
    "shortlist_preview": shortlist,
    "rejections_preview": rejections[:100],
    "inverter_limits": {
        "max_vdc": inv_max_vdc,
        "mppt_low": inv_mppt_low,
        "mppt_high": inv_mppt_high,
        "max_input_current_a": inv_max_input_current_a,
        "max_pv_input_power_w": inv_max_pv_input_power_w,
    },
    "pv_target_kwp": pv_kwp_final_target,
})


pv_state = {
    "meta": {
        "generated_at": datetime.now().isoformat(),
        "submission_key": latest_key,
        "project_root": project_root,
        "resolution_source": resolution_source,
        "source_stage_01": pv_stage_01_path,
        "source_inverter_state": inverter_state_path,
        "source_pv_schema": PV_PATH,
    },
    "selection_inputs": {
        "selected_inverter_identifier": inv_identifier,
        "inverter_rated_power_kw": inv_rated_power_kw,
        "inverter_max_vdc": inv_max_vdc,
        "inverter_mppt_low": inv_mppt_low,
        "inverter_mppt_high": inv_mppt_high,
        "inverter_max_input_current_a": inv_max_input_current_a,
        "inverter_max_pv_input_power_w": inv_max_pv_input_power_w,
        "pv_kwp_raw": pv_kwp_raw,
        "pv_kwp_final_target": pv_kwp_final_target,
    },
    "selected_pv": selected_pv,
    "shortlist": shortlist,
}


save_json(pv_state_path, pv_state)


if selected_pv:
    report_text = f"""# TENET Design Step 03 — PV Selection Report


## Target
- Final PV target size: **{pv_kwp_final_target:.4f} kWp**
- Raw PV size before margin: **{pv_kwp_raw:.4f} kWp**
- Selected inverter: **{inv_identifier}**


## Selected PV Module
- Identifier: **{selected_pv.get('identifier','')}**
- Module power: **{n(selected_pv.get('p_w'),0):.2f} W**
- Voc: **{n(selected_pv.get('voc'),0):.2f} V**
- Vmp: **{n(selected_pv.get('vmp'),0):.2f} V**
- Isc: **{n(selected_pv.get('isc'),0):.2f} A**
- Imp: **{n(selected_pv.get('imp'),0):.2f} A**


## Selected Array Design
- Modules per string: **{int(n(selected_pv.get('modules_per_string'),0))}**
- Parallel strings: **{int(n(selected_pv.get('num_strings'),0))}**
- Total modules: **{int(n(selected_pv.get('total_modules'),0))}**
- Total PV size: **{n(selected_pv.get('total_kwp'),0):.4f} kWp**
- Array Isc: **{n(selected_pv.get('array_isc'),0):.2f} A**
- Cold string Voc: **{n(selected_pv.get('cold_string_voc'),0):.2f} V**
- Operating string Vmp: **{n(selected_pv.get('operating_string_vmp'),0):.2f} V**
- Absolute error from target: **{n(selected_pv.get('kwp_abs_error'),0):.4f} kWp**
"""
else:
    report_text = "# TENET Design Step 03 — PV Selection Report\n\nNo compliant PV product/configuration found.\n"


with open(pv_report_md_path, "w", encoding="utf-8") as f:
    f.write(report_text)


# -------------------------------------------------
# PRINT
# -------------------------------------------------
if selected_pv:
    print("Selected PV:", selected_pv.get("identifier"))
    print("Modules per string:", int(n(selected_pv.get("modules_per_string"), 0)))
    print("Parallel strings:", int(n(selected_pv.get("num_strings"), 0)))
    print("Total modules:", int(n(selected_pv.get("total_modules"), 0)))
    print("Total PV size (kWp):", round(n(selected_pv.get("total_kwp"), 0), 4))
    print("Absolute error from target (kWp):", round(n(selected_pv.get("kwp_abs_error"), 0), 4))
else:
    print("No compliant PV configuration found.")
    print("Check debug file:", pv_debug_path)


print()
print("Saved:")
print(pv_state_path)
print(pv_report_md_path)
print(pv_debug_path)

```
